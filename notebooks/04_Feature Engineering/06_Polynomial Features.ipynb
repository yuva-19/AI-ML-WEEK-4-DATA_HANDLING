# POLYNOMIAL FEATURES

Polynomial features create new features from existing numerical features by generating **powers and interaction terms**.

They can help **linear models learn non-linear relationships**.

---

# 1. INTERACTION TERMS

Interaction terms represent the **combined effect of two or more features**.

Example:

```text
Age × Salary
````

Create manually:

```python
df["Age_Salary"] = (
    df["Age"] * df["Salary"]
)
```

`PolynomialFeatures` can automatically generate interaction and polynomial terms.

### Python

```python
from sklearn.preprocessing import PolynomialFeatures

poly = PolynomialFeatures(
    degree=2,
    include_bias=False
)

X_poly = poly.fit_transform(
    df[["Age", "Salary"]]
)
```

For two features and `degree=2`, the generated features include:

```text
Age
Salary
Age²
Age × Salary
Salary²
```

### Key Point

```text
Interaction Term → Combined effect of features

A × B
```

---

# 2. DEGREE SELECTION

The **degree** determines the complexity of the polynomial features.

### Degree 1

Original features:

```text
Age
Salary
```

### Degree 2

Includes:

```text
Age
Salary
Age²
Age × Salary
Salary²
```

### Degree 3

Adds cubic terms and higher-order interactions:

```text
Age³
Age² × Salary
Age × Salary²
Salary³
```

### General Rule

```text
Higher Degree
      ↓
More Features
      ↓
More Complexity
```

Usually start with **degree 2** and increase the degree only when validation performance justifies it.

---

# 3. NUMBER OF FEATURES

Polynomial features can increase the number of features very quickly.

For `n` input features and polynomial degree `d`, the number of generated terms can grow rapidly as the degree increases.

For example, with 2 features:

```text
Degree 1 → 2 features
Degree 2 → 5 features
Degree 3 → 9 features
```

With many input features, high polynomial degrees can create a very large feature space.

This can increase:

* Computational cost
* Memory usage
* Model complexity
* Overfitting risk

---

# 4. OVERFITTING RISK

High-degree polynomial features can make the model too complex.

This can cause **overfitting**.

### Overfitting

```text
Training Performance
        ↓
     Very High

Validation/Test Performance
        ↓
       Poor
```

The model learns the training data too closely instead of learning patterns that generalize to unseen data.

### Degree and Overfitting

```text
Low Degree
    ↓
Less Complexity
    ↓
Lower Overfitting Risk
```

```text
High Degree
    ↓
More Complexity
    ↓
Higher Overfitting Risk
```

---

# 5. DEGREE SELECTION USING VALIDATION

Do not simply choose the highest possible degree.

Compare different degrees using **validation or cross-validation**.

Example:

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline

model = make_pipeline(
    PolynomialFeatures(
        degree=2,
        include_bias=False
    ),
    LinearRegression()
)

model.fit(X_train, y_train)
```

Try different degrees:

```text
Degree 1
Degree 2
Degree 3
Degree 4
```

Then compare their validation or cross-validation performance.

Choose a degree that provides good generalization rather than simply the best training performance.

---

# 6. POLYNOMIAL FEATURES WITH LINEAR REGRESSION

Polynomial features are especially useful with models such as **Linear Regression**.

Example:

Original relationship:

```text
y = β₀ + β₁x
```

This can only represent a straight-line relationship.

After creating a squared feature:

```text
x²
```

the model can learn:

```text
y = β₀ + β₁x + β₂x²
```

This can represent a curved relationship.

The model is still **linear in its coefficients**, even though the relationship between `x` and `y` is non-linear.

---

# IMPORTANT DISTINCTION

Polynomial features do **not automatically make the model itself non-linear**.

They create transformed features such as:

```text
x
x²
x³
x₁ × x₂
```

A model such as Linear Regression can then use these features to learn a **non-linear relationship between the original input and target**.

### Example

```text
Original Features
       ↓
Polynomial Transformation
       ↓
x, x², x³, x₁×x₂
       ↓
Linear Regression
       ↓
Non-linear relationship
```

---

# KEY POINTS 🧠

```text
Interaction Term
→ Captures combined effect of features
```

```text
Degree
→ Controls polynomial complexity
```

```text
Higher Degree
→ More features
→ More complexity
→ Higher overfitting risk
```

```text
Polynomial Features
→ Create powers + interactions
→ Allow linear models to learn non-linear relationships
```

```text
Degree Selection
→ Use validation / cross-validation
→ Do not simply choose the highest degree
```

### Remember

> **Polynomial Features = Powers + Interaction Terms**

> **Higher Degree = More Complexity = Higher Overfitting Risk**

> **Polynomial features don't automatically make the model non-linear; they create non-linear features that allow models such as Linear Regression to learn non-linear relationships.**

```
```
