# Feature Creation



Feature engineering means **creating new useful features from existing data** to help with analysis and machine learning.

### General Workflow

```text
Raw Data
   ↓
Feature Engineering
   ↓
New Useful Features
   ↓
Better Analysis / ML Models
````

---

# 1. RATIO FEATURES

A **ratio feature** compares two variables by dividing one variable by another.

### General Form

```text
A / B
```

### Example: Salary per Year of Experience

```python
df["Salary_per_Year"] = (
    df["Salary"] / df["Experience"]
)
```

This creates a new feature representing the salary relative to experience.

### Example: BMI

BMI compares body weight with height.

```python
df["BMI"] = (
    df["Weight_kg"] /
    (df["Height_m"] ** 2)
)
```

### Purpose

Ratio features represent **relative relationships** between variables.

Common examples:

* Salary / Experience
* Profit / Revenue
* Debt / Income
* Weight / Height²
* Sales / Customers

### Important

Check for **division by zero** before creating ratio features.

Example:

```python
df["Salary_per_Year"] = (
    df["Salary"] /
    df["Experience"].replace(0, pd.NA)
)
```

# 2. COUNT FEATURES

A **count feature** represents how many items, events, or occurrences are associated with a record.

### Example: Number of Orders per Customer

```python
df["Order_Count"] = (
    df.groupby("Customer_ID")["Order_ID"]
    .transform("count")
)
```

This creates a feature containing the total number of orders associated with each customer.

### Count Missing / Non-Missing Values

Count how many contact details are available:

```python
df["Contact_Count"] = df[
    ["Phone", "Email", "Address"]
].notna().sum(axis=1)
```

Example:

```text
Phone   Email   Address   Contact_Count
Yes     Yes     NaN       2
Yes     NaN     Yes       2
NaN     NaN     Yes       1
```

### Common Count Features

* Purchase count
* Login count
* Transaction count
* Number of children
* Number of contacts
* Number of products purchased
* Number of previous interactions

Count features can help represent **frequency or activity level**.

---

# 3. INTERACTION FEATURES

An **interaction feature** combines two or more existing features to represent their **joint effect**.

A common interaction is multiplication:

```text
A × B
```

### Example: Total Spending

```python
df["Total_Spending"] = (
    df["Price"] * df["Quantity"]
)
```

Here, `Price` and `Quantity` interact to produce total spending.

### Another Example

```python
df["Age_Income"] = (
    df["Age"] * df["Income"]
)
```

This creates a feature representing the combined relationship between age and income.

### Ratio vs Interaction

| Feature Type | Operation | Example             |
| ------------ | --------- | ------------------- |
| Ratio        | `A / B`   | Salary / Experience |
| Interaction  | `A × B`   | Price × Quantity    |
| Difference   | `A - B`   | Income - Expenses   |

Interaction features can help models capture relationships that individual features may not represent on their own.

# 4. TIME-BASED FEATURES

Datetime values can be converted into useful features for analysis and machine learning.

Example:

```python
df["Year"] = df["Date"].dt.year

df["Month"] = df["Date"].dt.month

df["Day"] = df["Date"].dt.day

df["DayOfWeek"] = df["Date"].dt.dayofweek

df["Hour"] = df["Date"].dt.hour
```

### Common Time-Based Features

* Year
* Month
* Day
* Day of week
* Hour
* Minute
* Quarter
* Weekend indicator
* Business-hour indicator

---

## Weekend Feature

Create a Boolean feature indicating whether the date falls on a weekend.

```python
df["Is_Weekend"] = (
    df["Date"].dt.dayofweek >= 5
)
```

`dayofweek` uses:

```text
Monday    → 0
Tuesday   → 1
Wednesday → 2
Thursday  → 3
Friday    → 4
Saturday  → 5
Sunday    → 6
```

Therefore:

```text
0–4 → Weekday
5–6 → Weekend
```

---

## Business-Hour Feature

Create a feature indicating whether a timestamp falls within a specified business-hour range.

```python
df["Is_Business_Hour"] = (
    df["Date"].dt.hour.between(9, 17)
)
```

This returns `True` when the hour is between 9 and 17.

The exact business-hour range should depend on the problem and domain.

---

# IMPORTANT

Not every new feature is automatically useful.

A feature should be:

* Relevant to the problem
* Based on meaningful relationships
* Calculated correctly
* Available at the time of prediction
* Free from data leakage

### Data Leakage

Do not create a feature using information that would **not be available when the prediction is actually made**.

Example:

If predicting whether a customer will cancel an order, using information recorded **after the cancellation** would create data leakage.

# SUMMARY

| Feature Type    | Meaning                             | Example           |
| --------------- | ----------------------------------- | ----------------- |
| **Ratio**       | Relative relationship               | `A / B`           |
| **Count**       | Number of occurrences               | Number of orders  |
| **Interaction** | Joint effect of variables           | `A × B`           |
| **Time-Based**  | Information extracted from datetime | Year, Month, Hour |

### 🧠 Remember

```text
Ratio       → A / B
Count       → Number of occurrences
Interaction → A × B
Time-Based  → Year / Month / Day / Hour / DayOfWeek
```
