# LABEL ENCODING

**Label Encoding** converts each categorical value into a unique integer.

Example:

```text
Chennai     → 0
Madurai     → 1
Coimbatore  → 2
````

### Python

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df["City_encoded"] = le.fit_transform(df["City"])
```

`LabelEncoder` assigns an integer to each unique category.

---

## Example

Original:

| City       |
| ---------- |
| Chennai    |
| Madurai    |
| Coimbatore |
| Chennai    |

After Label Encoding:

| City       | City_encoded |
| ---------- | -----------: |
| Chennai    |            0 |
| Madurai    |            1 |
| Coimbatore |            2 |
| Chennai    |            0 |

The exact integer assigned to each category depends on the categories and the encoder.

---

## Advantages

* Simple and easy to implement
* Converts categories into numerical values
* Uses only one encoded column
* Memory efficient compared with one-hot encoding
* Can be useful for ordinal variables
* Can work well with some tree-based models

---

## Limitations

## 1. Can Introduce Artificial Ordering

Example:

```text
Chennai     → 0
Madurai     → 1
Coimbatore  → 2
```

The numerical values may create an artificial order:

```text
Coimbatore > Madurai > Chennai
```

But cities do not have a natural order.

---

## 2. Numerical Distance Becomes Meaningless

The model may interpret:

```text
Chennai → 0
Madurai → 1
Coimbatore → 2
```

as if:

```text
Distance(Chennai, Madurai) = 1
Distance(Madurai, Coimbatore) = 1
Distance(Chennai, Coimbatore) = 2
```

But there is no meaningful numerical distance between these categories.

---

## When to Use Label Encoding

Label encoding is safer when the categories have a **meaningful order**.

### Ordinal Variable

Example:

```text
Low < Medium < High
```

Can be encoded as:

```text
Low    → 0
Medium → 1
High   → 2
```

Here the numerical order represents the actual order:

```text
0 < 1 < 2
```

---

## Nominal vs Ordinal

### Nominal Categories

Categories have **no natural order**.

Examples:

```text
City
Color
Country
Department
```

For example:

```text
Chennai
Madurai
Coimbatore
```

Usually **avoid Label Encoding** for these categories.

### Ordinal Categories

Categories have a **meaningful order**.

Examples:

```text
Low < Medium < High
Small < Medium < Large
Poor < Average < Good
```

Label/integer encoding can make sense because the numerical order represents the actual category order.

---

## Key Rule 🧠

```text
Nominal
↓
No meaningful order
↓
Usually avoid Label Encoding
```

```text
Ordinal
↓
Meaningful order
↓
Integer/Label Encoding can make sense
```

> **Label Encoding is best suited to categorical variables where the numerical ordering represents a real order.**

### Remember

**Nominal → Usually avoid Label Encoding**

**Ordinal → Label Encoding can be appropriate**

# One Hot Encoding

**One-Hot Encoding** converts each category into a **separate binary column**.

Each category is represented using:

```text
1 → Category is present
0 → Category is not present
````

### Example

Original `City` column:

```text
Chennai
Madurai
Coimbatore
```

After One-Hot Encoding:

```text
City_Chennai
City_Madurai
City_Coimbatore
```

Representation:

```text
Chennai     → [1, 0, 0]
Madurai     → [0, 1, 0]
Coimbatore  → [0, 0, 1]
```

---

## Pandas

For a single column:

```python
pd.get_dummies(df["City"])
```

For a DataFrame:

```python
df = pd.get_dummies(
    df,
    columns=["City"]
)
```

Example:

```text
Original:

| City |
|---|
| Chennai |
| Madurai |
| Coimbatore |
```

After encoding:

```text
| City_Chennai | City_Madurai | City_Coimbatore |
|---:|---:|---:|
| 1 | 0 | 0 |
| 0 | 1 | 0 |
| 0 | 0 | 1 |
```

---

## Why Use One-Hot Encoding?

One-Hot Encoding avoids the **artificial numerical ordering** problem of Label Encoding.

### Label Encoding

```text
Chennai     → 0
Madurai     → 1
Coimbatore  → 2
```

A model may incorrectly interpret:

```text
Coimbatore > Madurai > Chennai
```

and may treat the numerical differences as meaningful.

### One-Hot Encoding

```text
Chennai     → [1, 0, 0]
Madurai     → [0, 1, 0]
Coimbatore  → [0, 0, 1]
```

Each category is represented independently.

There is **no artificial ordering or numerical distance** between categories.

---

## When to Use One-Hot Encoding

One-Hot Encoding is commonly used for **nominal categorical variables** where categories have no meaningful order.

Examples:

* City
* Country
* Color
* Department
* Product Type
* Payment Method

It can also be used for ordinal variables when preserving order is not required, but ordinal encoding may be more appropriate when the order itself carries useful information.

---

## SPARSE MATRICES

One-Hot Encoding can create a large number of columns when a categorical feature has many unique categories.

Example:

```text
[0, 0, 0, 1, 0, 0, 0, 0]
```

Most values are `0`, while only a small number of values are `1`.

This is called **sparse data**.

### Why Sparse Representation?

Instead of storing every zero, a sparse matrix can store the **non-zero values and their positions**.

This can significantly reduce memory usage when there are many categories and most entries are zero.

### Scikit-Learn

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(
    sparse_output=True
)
```

`sparse_output=True` produces a sparse matrix.

Sparse representation is especially useful for **high-cardinality categorical features**.

---

## DUMMY VARIABLE TRAP

Suppose `Gender` has two categories:

```text
Male
Female
```

One-Hot Encoding produces:

```text
| Male | Female |
|---:|---:|
| 1 | 0 |
| 0 | 1 |
| 1 | 0 |
```

Because there are only two categories:

```text
Female = 1 - Male
```

Therefore, one column can be perfectly predicted from the other.

For **linear models with an intercept**, this creates **perfect multicollinearity**.

This situation is called the **Dummy Variable Trap**.

---

## Avoiding the Dummy Variable Trap

Remove one dummy column.

Using Pandas:

```python
pd.get_dummies(
    df["Gender"],
    drop_first=True
)
```

For `K` categories:

```text
K categories → K - 1 dummy variables
```

Example:

```text
3 categories
    ↓
2 dummy variables
```

The omitted category becomes the **reference category**.

This is called the **reference-category approach**.

---

## Important

The dummy variable trap is mainly relevant when using **linear models with an intercept**, where perfect multicollinearity causes problems.

For many other machine learning models, such as tree-based models, this specific issue is generally not a concern.

---

## ONE-HOT ENCODING WITH UNKNOWN CATEGORIES

When using `OneHotEncoder` in a machine learning pipeline, new categories may appear in test or future data that were not present during training.

To handle them:

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(
    handle_unknown="ignore"
)
```

This prevents an error when an unseen category appears during transformation.

---

## QUICK COMPARISON

| Method               | Representation            | Best Use           |
| -------------------- | ------------------------- | ------------------ |
| **Label Encoding**   | Category → Integer        | Ordinal categories |
| **One-Hot Encoding** | Category → Binary columns | Nominal categories |

### Label Encoding

```text
Chennai     → 0
Madurai     → 1
Coimbatore  → 2
```

### One-Hot Encoding

```text
Chennai     → [1, 0, 0]
Madurai     → [0, 1, 0]
Coimbatore  → [0, 0, 1]
```

---

## 🧠 KEY RULE

```text
Nominal categorical data
        ↓
One-Hot Encoding
```

```text
Ordinal categorical data
        ↓
Ordinal / Integer Encoding
```

> **One-Hot Encoding represents each category independently using binary columns, avoiding artificial numerical ordering.**

# ADVANCED ENCODING METHODS

Advanced encoding methods are techniques used to convert categorical variables into numerical representations, especially when simple methods such as Label Encoding or One-Hot Encoding are not ideal.

Main methods:

1. Ordinal Encoding
2. Target Encoding
3. Frequency Encoding
4. Binary Encoding

---

## 1. ORDINAL ENCODING

**Ordinal Encoding** is used when categories have a **meaningful natural order**.

Example:

```text
Low    → 0
Medium → 1
High   → 2
````

The numerical values represent the actual order:

```text
Low < Medium < High
```

### Python

```python
from sklearn.preprocessing import OrdinalEncoder

encoder = OrdinalEncoder(
    categories=[["Low", "Medium", "High"]]
)

df["Satisfaction_encoded"] = encoder.fit_transform(
    df[["Satisfaction"]]
)
```

### Examples

```text
Low < Medium < High
```

```text
School < Bachelor < Master < PhD
```

```text
Small < Medium < Large
```

### When to Use

Use Ordinal Encoding when:

* Categories have a meaningful order.
* The order should be represented numerically.
* The distance between categories is acceptable for the intended model.

### Important

Do not use ordinal encoding simply because categories can be arranged alphabetically.

For example:

```text
Chennai
Madurai
Coimbatore
```

do not have a meaningful natural order.

### 🧠 Memory

> **Ordinal = Order**

---

## 2. TARGET ENCODING

**Target Encoding** replaces each category with a statistic calculated from the **target variable**.

For classification, the statistic is commonly the **mean target value**.

Example:

```text
City        Purchased
Chennai     1
Chennai     1
Chennai     0
Madurai     0
Madurai     0
```

Mean target:

```text
Chennai     → 0.67
Madurai     → 0.00
```

### Basic Idea

```text
Category
   ↓
Calculate target statistic
   ↓
Replace category with statistic
```

### Python

```python
target_mean = df.groupby("City")["Purchased"].mean()

df["City_target_encoded"] = df["City"].map(
    target_mean
)
```

### Example

```text
Chennai     → 0.67
Madurai     → 0.00
```

Here, the encoded value represents the average target value associated with each category.

### Useful For

* High-cardinality categorical features
* Categories where One-Hot Encoding would create too many columns
* Problems where the category has a strong relationship with the target

### Important: Data Leakage ⚠️

Target Encoding can cause **data leakage** if target statistics are calculated using validation or test data.

Incorrect approach:

```text
Entire Dataset
      ↓
Calculate Target Statistics
      ↓
Split into Train/Test
```

The target information from the test set has already influenced the encoding.

Correct general approach:

```text
Training Data
     ↓
Calculate Target Statistics
     ↓
Encode Training Data

Validation/Test Data
     ↓
Use Training Statistics
     ↓
Encode Validation/Test Data
```

Target encoding should be performed carefully, especially when building ML pipelines.

### 🧠 Memory

> **Target Encoding = Category → Target Statistic**

---

## 4. BINARY ENCODING

**Binary Encoding** converts categories into integers and then represents those integers using **binary digits**.

### Process

```text
Category
   ↓
Integer
   ↓
Binary Representation
   ↓
Binary Columns
```

Example:

```text
A → 1 → 001
B → 2 → 010
C → 3 → 011
D → 4 → 100
```

The binary digits are then represented as separate columns.

Example:

```text
| Bit_1 | Bit_2 | Bit_3 |
|------:|------:|------:|
| 0     | 0     | 1     |
| 0     | 1     | 0     |
| 0     | 1     | 1     |
| 1     | 0     | 0     |
```

### Why Use Binary Encoding?

One-Hot Encoding creates one column for **every category**.

For example:

```text
10 categories
    ↓
10 One-Hot columns
```

Binary Encoding requires approximately:

```text
log₂(number of categories)
```

binary columns.

For example:

```text
8 categories
↓
log₂(8)
↓
3 binary columns
```

Therefore, Binary Encoding can create **far fewer columns than One-Hot Encoding** for high-cardinality categorical variables.

### Useful For

* High-cardinality categorical variables
* Reducing dimensionality compared with One-Hot Encoding
* Situations where One-Hot Encoding would create too many columns

### Important

Binary Encoding introduces a numerical representation of categories, but it generally uses fewer dimensions than One-Hot Encoding.

### 🧠 Memory

> **Binary Encoding = Category → Integer → Binary**

---

# QUICK COMPARISON

| Encoding               | Basic Idea                  | Best Use                    |
| ---------------------- | --------------------------- | --------------------------- |
| **Ordinal Encoding**   | Category → Ordered Integer  | Ordinal categories          |
| **Target Encoding**    | Category → Target Statistic | High-cardinality categories |
| **Frequency Encoding** | Category → Frequency        | High-cardinality categories |
| **Binary Encoding**    | Category → Integer → Binary | High-cardinality categories |

---

# ENCODING DECISION

```text
Does the category have a meaningful order?
        ↓
      YES
        ↓
Ordinal Encoding
```

```text
Nominal + Low Cardinality
        ↓
One-Hot Encoding
```

```text
Nominal + High Cardinality
        ↓
Consider:
Target Encoding
Frequency Encoding
Binary Encoding
```

### Important

Target Encoding requires special care because it uses the **target variable** and can cause data leakage.

---

# 🧠 QUICK MEMORY

```text
Ordinal
→ Order
→ Low < Medium < High
```

```text
Target
→ Target statistic
→ Category → Target relationship
```

```text
Frequency
→ How common
→ Category → Count / Proportion
```

```text
Binary
→ Integer → Binary
→ Fewer columns for high cardinality
```

> **Choose the encoding method based on the type of categorical variable, its cardinality, whether an order exists, and the requirements of the ML model.**
