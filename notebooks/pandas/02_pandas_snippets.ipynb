Series:
- A one-dimensional labeled array.
- Holds data of any type (int, float, string, etc.).
- Similar to a column in Excel or a single list with labels (index).

DataFrame:
- A two-dimensional labeled data structure.
- Consists of rows and columns (like a table).
- Each column is a Series, combined together into a DataFrame.
- Can be created from dictionaries, lists, CSV, Excel, SQL, etc.

# Filtering - Using Conditionals to Filter Rows and Columns

```python
people = {
    "first": ['Corey', 'Jane', 'John'],
    "last": ['Schafer', 'Doe', 'Doe'],
    "email": ['CoreyMSchafer@gmail.com', 'JaneDoe@email.com', 'JohnDoe@email.com']
}
```

```python
import pandas as pd
df=pd.DataFrame(people)
df
```

```python
filt = (df['last'] == 'Doe')#used to filter the specific data like country,area,role etc..
```

```python
df[filt]
```

```python
df.loc[filt, 'email']
```

```python
# We can also use and or not operators here
filt = (df['last'] == 'Doe') & (df['first'] == 'John')#AND operator
```

```python
df.loc[filt]
```

```python
filt = (df['last'] == 'Schafer') | (df['first'] == 'John')#OR operator
```

```python
df.loc[filt,'email']
```

```python
df.loc[~filt,'email']#NOT operator gives opossite of filt
```

# Updating Rows and Columns - Modifying Data Within DataFrames

```python
people = {
    "first": ['Corey', 'Jane', 'John'],
    "last": ['Schafer', 'Doe', 'Doe'],
    "email": ['CoreyMSchafer@gmail.com', 'JaneDoe@email.com', 'JohnDoe@email.com']
}
```

```python
import pandas as pd
df=pd.DataFrame(people)
df
```

```python
#to change every columns
df.columns=['first name','last name','email']
df
```

```python
df.columns = [x.upper() for x in df.columns] #to update columns(upper lower capitalize)
df
```

```python
#to change the string
df.columns=df.columns.str.replace(' ', '_')
df
```

```python
df.columns = [x.lower() for x in df.columns] #to update columns(upper lower capitalize)
df
```

```python
#to change particular columns instead of everything
df.rename(columns={'first_name':'first','last_name':'last'},inplace=True)
```

```python
df
```

```python
# to update the particular row
df.loc[2] = ['John','Smith','JohnSmith@email.com']
```

```python
df
```

```python
#to update the specific things in a row
df.loc[2,['last','email']] = ['Doe','JohnDoe@email.com']
df
```

```python
#to change single value alone
df.loc[2,'last']='Smith'
df
```

```python
df.loc[2,'last']='Doe'
df
```

```python
# if u wanna filter out and update the things
filt = (df['email'] == 'JohnDoe@email.com')
df.loc[filt,'last'] = 'Smith'
df
```

```python
#to update every columns
df['email'] = df['email'].str.lower()
df
```

# Sorting Data

```python
people = {
    "first": ['Corey', 'Jane', 'John','Adam'],
    "last": ['Schafer', 'Doe', 'Doe','Doe'],
    "email": ['CoreyMSchafer@gmail.com', 'JaneDoe@email.com', 'JohnDoe@email.com','AdamDoe@email.com']
}
```

```python
import pandas as pd
df=pd.DataFrame(people)
df
```

```python
df.sort_values(by='last', ascending=False)
```

```python
# Sort by Multiple rows
df.sort_values(by=['last', 'first'], ascending=False)
```

```python
# Ascending one value and descending one value
# False=Descending and True=Ascending
df.sort_values(by=['last', 'first'], ascending=[False,True],inplace=True)
df
```

```python
# To reset everything
df=df.sort_index()
df
```

# Cleaning Data - Casting Datatypes and Handling Missing Values

```python
import pandas as pd
import numpy as np
```

```python
people = {
    'first': ['Corey', 'Jane', 'John', 'Chris', np.nan, None, 'NA'],
    'last': ['Schafer', 'Doe', 'Doe', 'Schafer', np.nan, np.nan, 'Missing'],
    'email': ['CoreyMSchafer@gmail.com', 'JaneDoe@email.com', 'JohnDoe@email.com', None, np.nan, np.nan, 'Missing'],
    'age': ['33', '55', '63', '36', None, None, 'Missing']
}
```

```python
dfb=pd.DataFrame(people)
```

```python
dfb
```

```python
# it removes row when it have atleast one NaN(np.nan)
dfb.dropna()
```

```python
# removes a row when any value has NaN
# index used to remove the rows
dfb.dropna(axis= 'index', how = 'any')
```

```python
# removes a row when all value are NaN
dfb.dropna(axis= 'index', how = 'all')
```

```python
# if all the columns are Nan or None it will return nothing
# columns used to remove the column
dfb.dropna(axis= 'columns', how = 'all')
```

```python
# if any of  the column are  fully Nan or None it will return nothing
#  4 and 5 full of Nan so it returns nothing
dfb.dropna(axis= 'columns', how = 'any')
```

```python
# Now if we want to clean data like we only need email for anlaysis
# name is not needed
# using the above method if name has None or NaN-->email will be also removed if present thats valuable data for us
# So we Using
dfb.dropna(axis= 'index', how = 'any',subset=['email'])
```

```python
# drop row when all the values are nan or none from last and email
dfb.dropna(axis= 'index', how = 'all',subset=['last','email'])
```

```python
import pandas as pd
import numpy as np
```

```python
people = {
    'first': ['Corey', 'Jane', 'John', 'Chris', np.nan, None, 'NA'],
    'last': ['Schafer', 'Doe', 'Doe', 'Schafer', np.nan, np.nan, 'Missing'],
    'email': ['CoreyMSchafer@gmail.com', 'JaneDoe@email.com', 'JohnDoe@email.com', None, np.nan, np.nan, 'Missing'],
    'age': ['33', '55', '63', '36', None, None, 'Missing']
}
```

```python
# To get better understanding
df.fillna('MISSING')
```

# merge join and concate the dataframes

```python
import pandas as pd


df1 = pd.DataFrame({
    "FellowshipID": [1001, 1002, 1003, 1004],
    "FirstName": ["Frodo", "Samwise", "Gandalf", "Pippin"],
    "Skills": ["Hiding", "Gardening", "Spells", "Fireworks"]
})

df2 = pd.DataFrame({
    "FellowshipID": [1001, 1002, 1006, 1007, 1008],
    "FirstName": ["Frodo", "Samwise", "Legolas", "Elrond", "Boromir"],
    "Age": [50, 39, 2931, 6520, 51]
})
```

```python
df1
```

```python
df2
```

```python
# This gonna give inner join-->intersection values alone
# 1001 and 1002 are only things same here so it give those as output
df1.merge(df2)
```

```python
df1.merge(df2,how ='inner', on =['FellowshipID'])
```

```python
df1.merge(df2,how ='inner', on =['FellowshipID','FirstName'])
```

```python
df1.merge(df2, how= 'outer')
```

```python
df1.merge(df2, how= 'left')
# This takes left side(df1) and returns the df1 alone but it additionaly adds the data from df2 if any presents
# In this case df2 have age so it adds it
# 1001 and 1002 are present in df2 and we have age values so its adds it 
# 1003 and  1004 not present in df2 so it returns as NaN
```

```python
# it does the same but with right data(df2) and it compares with left(df1)
df1.merge(df2, how= 'right')
```

```python
# do compare every single value
# take one from df1 and compare withe every df2
# take next form df1 and again compare with every df2
df1.merge(df2, how= 'cross')
```

```python
df1.join(df2, on='FellowshipID', how = 'outer', lsuffix='_LEFT_', rsuffix = '_RIGHT_')
```

```python
# concate function
pd.concat([df1,df2], join ='inner')
```

```python
pd.concat([df1,df2], join ='outer')
```

```python
pd.concat([df1,df2], join ='outer',axis = 1)
```

# pivot and pivot tabels

```python
import pandas as pd

df = pd.DataFrame({
    "date": [
        "5/1/2017", "5/2/2017", "5/3/2017",
        "5/1/2017", "5/2/2017", "5/3/2017",
        "5/1/2017", "5/2/2017", "5/3/2017"
    ],
    "city": [
        "new york", "new york", "new york",
        "mumbai", "mumbai", "mumbai",
        "beijing", "beijing", "beijing"
    ],
    "temperature": [65, 66, 68, 75, 78, 82, 80, 77, 79],
    "humidity": [56, 58, 60, 80, 83, 85, 26, 30, 35]
})
```

```python
df
```

```python
df.pivot(index= 'date', columns= 'city')
```

```python
# if u want only humidity
df.pivot(index= 'date', columns= 'city',values = 'humidity')
```

```python
import pandas as pd

df = pd.DataFrame({
    "date": [
        "5/1/2017", "5/1/2017", "5/2/2017", "5/2/2017",
        "5/1/2017", "5/1/2017", "5/2/2017", "5/2/2017"
    ],
    "city": [
        "new york", "new york", "new york", "new york",
        "mumbai", "mumbai", "mumbai", "mumbai"
    ],
    "temperature": [65, 61, 70, 72, 75, 78, 82, 80],
    "humidity": [56, 54, 60, 62, 80, 83, 85, 26]
})
df
```

```python
df.pivot_table(index='city', columns= 'date')
```

```python
# to aggfunc()
df.pivot_table(index='city', columns= 'date',aggfunc='sum')
```

```python
df.pivot_table(index='city', columns= 'date',aggfunc='count')
```

```python
df.pivot_table(index='city', columns= 'date',aggfunc='diff')
```

```python
df.pivot_table(index='city', columns= 'date',margins=True)
# it does the mean to the rows
# and columns
# eg here humidity averaged for each city in row wise
# and column wise for each date
# and same for temperature
```
