# Seaborn 📊

> Seaborn is a Python visualization library built on top of Matplotlib.

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

## 1. Themes

Set the overall plot appearance.

```python
sns.set_theme(style="whitegrid")
```

Common styles:

`darkgrid`, `whitegrid`, `dark`, `white`, `ticks`

---

## 2. Statistical Plotting

### Scatter Plot

```python
sns.scatterplot(
    data=df,
    x="Experience",
    y="Salary"
)
```

### Histogram

```python
sns.histplot(
    data=df,
    x="Salary",
    kde=True
)
```

### Box Plot

```python
sns.boxplot(
    data=df,
    x="Department",
    y="Salary"
)
```

### Bar Plot

```python
sns.barplot(
    data=df,
    x="Department",
    y="Salary"
)
```

### Count Plot

```python
sns.countplot(
    data=df,
    x="Department"
)
```

---

## 3. Color Palettes

Set palette:

```python
sns.set_palette("Set2")
```

Use palette:

```python
sns.boxplot(
    data=df,
    x="Department",
    y="Salary",
    palette="Set2"
)
```

**Theme → Overall appearance**

**Palette → Plot colors**

---

## 4. Hue

Adds another categorical variable to a plot.

```python
sns.scatterplot(
    data=df,
    x="Experience",
    y="Salary",
    hue="Department"
)
```

Useful for **multivariate EDA**.

---

## 5. Axes-Level vs Figure-Level

### Axes-Level

Works with one Axes.

Examples:

`scatterplot()`

`histplot()`

`boxplot()`

`barplot()`

`countplot()`

Can pass an existing Axes:

```python
fig, ax = plt.subplots()

sns.scatterplot(
    data=df,
    x="Experience",
    y="Salary",
    ax=ax
)
```

### Figure-Level

Manages the entire Figure.

Examples:

`relplot()`

`displot()`

`catplot()`

```python
sns.relplot(
    data=df,
    x="Experience",
    y="Salary",
    hue="Department"
)
```

### Key Difference

**Axes-level → controls an individual plotting area**

**Figure-level → manages the overall visualization/Figure**

---

## 6. Pair Plot

Useful for pairwise relationships between numerical variables.

```python
sns.pairplot(df)
```

With categories:

```python
sns.pairplot(
    df,
    hue="Department"
)
```
