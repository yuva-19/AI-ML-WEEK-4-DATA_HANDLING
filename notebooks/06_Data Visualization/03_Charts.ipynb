````python
import pandas as pd

data = {
    "Month": [
        "Jan", "Feb", "Mar", "Apr", "May", "Jun",
        "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
    ],

    "Sales": [
        120, 135, 150, 142, 165, 180,
        175, 190, 210, 225, 240, 260
    ],

    "Profit": [
        18, 21, 25, 23, 29, 32,
        30, 35, 39, 43, 47, 52
    ],

    "Customers": [
        1200, 1300, 1450, 1380, 1550, 1700,
        1650, 1800, 1950, 2100, 2250, 2450
    ],

    "Department": [
        "Electronics", "Fashion", "Home", "Electronics",
        "Beauty", "Fashion", "Electronics", "Home",
        "Fashion", "Electronics", "Beauty", "Home"
    ],

    "Average_Order_Value": [
        45, 48, 51, 49, 55, 57,
        53, 59, 61, 64, 67, 70
    ]
}

df = pd.DataFrame(data)
````

````python
import seaborn as sns
import matplotlib.pyplot as plt
````

````python
# Task 1 — Line Plot

sns.lineplot(
    data = df,
    x ='Month',
    y = 'Sales'
)
````

````python
# Scatter Plot

sns.scatterplot(
    data = df,
    x = 'Customers',
    y = 'Sales'
)
````

````python
#  Histogram

sns.histplot(
    data = df,
    x ='Sales',
    kde = True
)
````

````python
# Bar Chart
sns.barplot(
    data =df,
    x = 'Department',
    y = 'Sales'
)
````

````python
# Count Plot
sns.countplot(
    data = df,
    y = 'Department'
)
````

````python
# Box Plot

sns.boxplot(
    data =df,
    x = 'Department',
    y = 'Sales'
)
````

````python
# Violin Plot
sns.violinplot(
    data =df,
    x = 'Department',
    y = 'Sales'
)
````

````python
# KDE
sns.kdeplot(
    data =df,
    x = 'Sales'
)
````

````python
sns.kdeplot(
    data=df,
    x="Sales",
    hue="Department"
)
````

````python
# ECDF
sns.ecdfplot(
    data=df,
    x='Sales'
)
````

````python
# Heatmap

corr = df.corr(numeric_only = True)

sns.heatmap(
    corr,
    annot = True
)
````

````python
# Pair Plot

sns.pairplot(df)
````

````python
sns.pairplot(
    df,
    x_vars=["Sales", "Profit"],
    y_vars=["Customers", "Average_Order_Value"]
)
````

````python
# Pie Chart

df['Department'].value_counts().plot(
    kind = 'pie',
    autopct = '%1.1f%%'
)
````


# Charts for EDA 📊

## Line Plot

Shows trends across an ordered variable/time.

```python
sns.lineplot(
    data=df,
    x="Month",
    y="Sales"
)
```

Use for: **time/order → numerical**

---

## Scatter Plot

Shows relationship between two numerical variables.

```python 
sns.scatterplot(
    data=df,
    x="Experience",
    y="Salary"
)
```

Use for: **numerical × numerical**

---

## Histogram

Shows distribution of a numerical variable using bins.

```python  
sns.histplot(
    data=df,
    x="Salary"
)
```

With KDE:

```python  
sns.histplot(
    data=df,
    x="Salary",
    kde=True
)
```

Use for: **numerical distribution**

---

## Bar Chart

Compares a numerical statistic across categories.

```python  
sns.barplot(
    data=df,
    x="Department",
    y="Salary"
)
```

Use for: **categorical → numerical summary**

---

## Count Plot

Shows the number of observations in each category.

```python  
sns.countplot(
    data=df,
    x="Department"
)
```

Use for: **categorical frequency / class imbalance**

---

## Pie Chart

Shows category proportions.

```python  
df["Department"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)
```

Use for: **category proportions**

Use sparingly; bar charts are usually easier to compare.

---

## Box Plot

Shows median, quartiles, spread and potential outliers.

```python  
sns.boxplot(
    data=df,
    x="Department",
    y="Salary"
)
```

Use for: **numerical distribution, especially across categories**

---

## Violin Plot

Combines box-plot information with distribution density.

```python 
sns.violinplot(
    data=df,
    x="Department",
    y="Salary"
)
```

Useful for comparing **distribution shapes across groups**.

---

## KDE Plot

Shows a smooth estimate of a numerical distribution.

```python 
sns.kdeplot(
    data=df,
    x="Salary"
)
```

With groups:

```python 
sns.kdeplot(
    data=df,
    x="Salary",
    hue="Department"
)
```

KDE = **Kernel Density Estimation**

---

## ECDF Plot

Shows the **cumulative proportion of observations ≤ each value**.

```python 
sns.ecdfplot(
    data=df,
    x="Salary"
)
```

Useful for:

- Cumulative distribution
- Percentile-type questions
- Comparing distributions

ECDF = **Empirical Cumulative Distribution Function**

---

## Heatmap

Uses colors to represent values in a matrix.

Common EDA use: **Correlation Matrix**

```python 
corr = df.corr(
    numeric_only=True
)

sns.heatmap(
    corr,
    annot=True
)
```

---

## Pair Plot

Shows pairwise relationships between multiple numerical variables.

```python 
sns.pairplot(df)
```

With categories:

```python 
sns.pairplot(
    df,
    hue="Department"
)
```

```text
Diagonal
→ Individual distributions

Off-diagonal
→ Pairwise relationships
```

---

# Chart Selection Guide

```text
Trend
→ Line Plot

Numerical × Numerical
→ Scatter Plot

Numerical Distribution
→ Histogram / KDE

Category Comparison
→ Bar Plot

Category Counts
→ Count Plot

Proportions
→ Pie Chart (sparingly)

Outliers + Distribution
→ Box Plot

Distribution + Density by Group
→ Violin Plot

Cumulative Distribution
→ ECDF

Correlation Matrix
→ Heatmap

Many Pairwise Relationships
→ Pair Plot
```
