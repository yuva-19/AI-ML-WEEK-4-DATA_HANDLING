```python
import pandas as pd

data = {
    "Month": [
        "Jan","Feb","Mar","Apr","May","Jun",
        "Jul","Aug","Sep","Oct","Nov","Dec"
    ],

    "Sales": [
        120,135,150,142,165,180,
        175,190,210,225,240,260
    ],

    "Profit": [
        18,21,25,23,29,32,
        30,35,39,43,47,52
    ],

    "Customers": [
        1200,1300,1450,1380,1550,1700,
        1650,1800,1950,2100,2250,2450
    ],

    "Department": [
        "Electronics","Fashion","Home","Electronics",
        "Beauty","Fashion","Electronics","Home",
        "Fashion","Electronics","Beauty","Home"
    ]
}

df = pd.DataFrame(data)
```

```python
# Task 1 — Colorblind-Friendly Visualization
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")
sns.set_palette("colorblind")

# Sales vs Profit line plot
plt.figure(figsize=(8,5))
sns.lineplot(x="Month", y="Sales", data=df, label="Sales")
sns.lineplot(x="Month", y="Profit", data=df, label="Profit")
plt.title("Sales vs Profit (Colorblind-Friendly)")
plt.legend()
plt.show()

# Department categorical plot
plt.figure(figsize=(8,5))
sns.countplot(x="Department", data=df)
plt.title("Department Distribution")
plt.show()
```

```python
# Task 2 — Annotation

plt.figure(figsize=(8,5))
sns.lineplot(x="Month", y="Sales", data=df, marker="o", color="blue")

# Find highest sales month
max_row = df.loc[df["Sales"].idxmax()]

plt.annotate(
    f"Highest Sales: {max_row['Sales']} ({max_row['Month']})",
    xy=(max_row["Month"], max_row["Sales"]),
    xytext=(max_row["Month"], max_row["Sales"]+20),
    arrowprops=dict(facecolor="black", arrowstyle="->")
)

plt.title("Monthly Sales with Annotation")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
# Task 3 — Dashboard Thinking

import matplotlib.pyplot as plt
import seaborn as sns

# --- KPI Calculations ---
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_customers = df["Customers"].sum()
avg_sales = df["Sales"].mean()

print("KPI 1 - Total Sales:", total_sales)
print("KPI 2 - Total Profit:", total_profit)
print("KPI 3 - Total Customers:", total_customers)
print("KPI 4 - Average Sales:", avg_sales)

# --- Dashboard Layout ---
fig, axes = plt.subplots(2, 2, figsize=(14,10))

# Monthly Sales Trend (Line plot)
sns.lineplot(x="Month", y="Sales", data=df, ax=axes[0,0], marker="o", color="blue")
axes[0,0].set_title("Monthly Sales Trend")
axes[0,0].set_xlabel("Month")
axes[0,0].set_ylabel("Sales")
axes[0,0].grid(True)

# Monthly Profit Trend (Line plot)
sns.lineplot(x="Month", y="Profit", data=df, ax=axes[0,1], marker="o", color="green")
axes[0,1].set_title("Monthly Profit Trend")
axes[0,1].set_xlabel("Month")
axes[0,1].set_ylabel("Profit")
axes[0,1].grid(True)

# Sales Distribution (Histogram)
sns.histplot(df["Sales"], bins=6, kde=True, ax=axes[1,0], color="purple")
axes[1,0].set_title("Sales Distribution")
axes[1,0].set_xlabel("Sales")
axes[1,0].set_ylabel("Frequency")

# Department Comparison (Bar chart)
sns.barplot(x="Department", y="Sales", data=df, ax=axes[1,1], estimator=sum, ci=None, palette="colorblind")
axes[1,1].set_title("Department Sales Comparison")
axes[1,1].set_xlabel("Department")
axes[1,1].set_ylabel("Total Sales")

plt.tight_layout()
plt.show()
```

```python
#  Task 4 — Plotly Interactive Scatter

import plotly.express as px
import plotly.io as pio

pio.renderers.default = "browser"  # change to "notebook" if using Jupyter

fig = px.scatter(
    df,
    x="Customers",
    y="Sales",
    hover_data=["Month", "Profit"],
    color="Department"
)

fig.update_layout(title="Interactive Sales vs Customers Scatter Plot")
fig.show()

```
