`````python
import pandas as pd

data = {
    "Month": [
        "Jan", "Feb", "Mar", "Apr", "May", "Jun",
        "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
    ],

    "Sales": [
        120, 135, 150, 142, 165, 180,
        175, 190, 210, 225, 240, 260
    ],

    "Profit": [
        20, 24, 28, 25, 31, 35,
        33, 39, 43, 47, 52, 58
    ],

    "Customers": [
        1200, 1300, 1450, 1380, 1550, 1700,
        1650, 1800, 1950, 2100, 2250, 2450
    ]
}

df = pd.DataFrame(data)
`````

`````python
import matplotlib.pyplot as plt
`````

`````python
fig = plt.figure(figsize=(8, 5))
plt.plot(df['Month'],df['Sales'],color = 'Blue')
plt.grid(True)
`````

`````python
plt.figure(figsize=(8,5))
plt.plot(df["Month"], df["Sales"], label="Sales", color="blue")
plt.plot(df["Month"], df["Profit"], label="Profit", color="green")

plt.title("Sales & Profit Trend")
plt.xlabel("Month")
plt.ylabel("Values")
plt.legend()
plt.grid(True)
plt.show()
`````

`````python
# fig, axes = plt.subplots(rows, columns, figsize=(width,height))
fig, axes = plt.subplots(2, 2, figsize=(12,8))

# Sales Trend
axes[0,0].plot(df["Month"], df["Sales"], color="blue")
axes[0,0].set_title("Sales Trend")
axes[0,0].set_xlabel("Month")
axes[0,0].set_ylabel("Sales")
axes[0,0].grid(True)

# Profit Trend
axes[0,1].plot(df["Month"], df["Profit"], color="green")
axes[0,1].set_title("Profit Trend")
axes[0,1].set_xlabel("Month")
axes[0,1].set_ylabel("Profit")
axes[0,1].grid(True)

# Customers Trend (Bar chart)
axes[1,0].bar(df["Month"], df["Customers"], color="orange")
axes[1,0].set_title("Customers Trend")
axes[1,0].set_xlabel("Month")
axes[1,0].set_ylabel("Customers")

# Sales vs Customers (Scatter plot)
axes[1,1].scatter(df["Sales"], df["Customers"], color="red")
axes[1,1].set_title("Sales vs Customers")
axes[1,1].set_xlabel("Sales")
axes[1,1].set_ylabel("Customers")
axes[1,1].grid(True)

plt.tight_layout()
plt.show()
`````

`````python
plt.figure(figsize=(8,5))
plt.plot(df["Month"], df["Sales"], 
         marker="o", linestyle="--", linewidth=2, color="purple", label="Sales")

plt.title("Styled Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.grid(True)
plt.show()
`````

# Matplotlib 📊

> Matplotlib is a Python library used for creating and customizing visualizations.

```python
import matplotlib.pyplot as plt
````

## Figure

> Figure = entire canvas containing plots.

```python
fig = plt.figure(
    figsize=(8, 5)
)
```

## Axes

> Axes = actual plotting area.

```python
fig, ax = plt.subplots()

ax.plot(x, y)

plt.show()
```

**Figure → Canvas**

**Axes → Plotting area**

**Axis → X-axis / Y-axis**

## Subplots

Create multiple plots in one Figure.

```python
fig, axes = plt.subplots(
    1,
    2,
    figsize=(10, 4)
)
```

2 × 2:

```python
fig, axes = plt.subplots(
    2,
    2,
    figsize=(10, 8)
)
```

Access:

```python
axes[0, 0]
axes[0, 1]
axes[1, 0]
axes[1, 1]
```

## Styling

```python
ax.set_title("Sales")

ax.set_xlabel("Month")

ax.set_ylabel("Revenue")

ax.grid()
```

Line styling:

```python
ax.plot(
    x,
    y,
    linestyle="--",
    marker="o",
    linewidth=2
)
```

## Legends

```python
ax.plot(
    x,
    y1,
    label="Sales"
)

ax.plot(
    x,
    y2,
    label="Profit"
)

ax.legend()
```

## Saving Figures

```python
plt.savefig(
    "plot.png",
    dpi=300,
    bbox_inches="tight"
)
```

Other formats:

`.png`, `.jpg`, `.pdf`, `.svg`

## Recommended Object-Oriented Approach

```python
fig, ax = plt.subplots()

ax.plot(x, y)

ax.set_title("My Plot")

ax.set_xlabel("X")

ax.set_ylabel("Y")

ax.legend()

ax.grid()

plt.tight_layout()

plt.show()
```

## Key Workflow

**Figure → Axes → Plot → Style → Labels → Legend → Layout → Save**
