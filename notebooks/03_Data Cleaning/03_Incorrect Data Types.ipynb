# INCORRECT DATA TYPES

## What is a Data Type?

A data type tells Python/Pandas what kind of value a column contains and how the value should be interpreted.

Examples:

| Column | Example | Appropriate Type |
|---|---|---|
| Age | `25` | Integer |
| Salary | `45000.50` | Float |
| Name | `"Yuva"` | String |
| Passed | `True` | Boolean |
| Date | `2026-08-12` | Datetime |
| Gender | `"Male"` | Category |

---

## What is an Incorrect Data Type?

An incorrect data type occurs when the stored data type does not match the actual meaning or nature of the data.

Example:

```text
Age:
20
25
30
35
````

If Pandas stores `Age` as `object` instead of a numerical type, the data type may be incorrect.

---

## Why Correct Data Types Matter?

Incorrect data types can cause:

* Calculation errors
* Incorrect sorting
* Higher memory usage
* Problems during data analysis
* Problems during visualization
* Problems when preparing data for Machine Learning

Example:

```python
"10" + "20"
```

Output:

```text
"1020"
```

Because both values are strings.

Correct numerical values:

```python
10 + 20
```

Output:

```text
30
```

---

## Detect Data Types

Use:

```python
df.dtypes
```

This shows the data type of every column.

---

## `df.info()`

Use:

```python
df.info()
```

It shows:

* Column names
* Non-null values
* Data types
* Memory usage

---

## Important: `object` Does NOT Always Mean Incorrect

`object` is commonly used by Pandas for text data.

Example:

```text
Name → object
```

This can be completely normal.

But:

```text
Age → object
```

may be incorrect if the column contains numerical values stored as strings.

Therefore:

> Always determine the correct data type based on the **meaning of the column**, not just the current Pandas dtype.

## Example: Numbers Stored as Strings

```python
    "Age": ["20", "25", "30", "35"]
}

df = pd.DataFrame(data)
```

Check the data type:

```python
df.dtypes
```

`Age` is stored as a string/object instead of a numerical type.

Convert it:

```python
df["Age"] = pd.to_numeric(df["Age"])
```

Now `Age` is treated as a numerical column.

---

## Example: Salary Stored as Text

Original data:

```text
₹30,000
₹45,000
₹50,000
```

The values contain the currency symbol and commas, so they cannot directly be treated as numerical values.

First remove `₹` and `,`:

```python
df["Salary"] = (
    df["Salary"]
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
)
```

Then convert to numeric:

```python
df["Salary"] = pd.to_numeric(df["Salary"])
```

Now the Salary column can be used for numerical calculations.

---

## Main Data Types

| Data Type    | Meaning                        | Example               |
| ------------ | ------------------------------ | --------------------- |

| **Float**    | Decimal numbers                | `10.5`, `25.75`       |
| **String**   | Text                           | `"Chennai"`, `"Yuva"` |

| **Datetime** | Date and time                  | `2026-08-12`          |
| **Category** | Limited/repeated set of values | `"Male"`, `"Female"`  |

# INTEGER

## What is an Integer?

An integer is a **whole number without a decimal part**.

Examples:

```text
-10
-1
0
1
25
100
5000
````

Not integers:

```text
10.5
25.75
3.14
```

These are **float** values.

---

## Integer in Pandas

A common Pandas integer dtype is:

```python
int64
```

Example:

```text
Age
---
20
25
30
35
```

---

## Numbers Stored as Strings

Sometimes numbers are stored as text.

Example:

```text
Age
---
"20"
"25"
"30"
"35"
```

Pandas may detect this as:

```text
object
```

or:

```text
string
```

These values look like numbers but are actually **strings**.

### Convert Clean Numeric Strings

```python
df["Age"] = df["Age"].astype(int)
```

After conversion, the values are treated as integers.

---

## Integer and Missing Values

Standard Pandas `int64` cannot normally store `NaN`.

Example:

```text
20
25
NaN
35
```

Trying:

```python
df["Age"].astype(int)
```

may produce an error because regular integers cannot represent `NaN`.

### Solution 1: Handle Missing Values First

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Age"] = df["Age"].astype(int)
```

Here:

1. Missing values are replaced using the median.
2. The column is converted to integer.

### Solution 2: Use Pandas Nullable Integer

Pandas provides a nullable integer dtype:

```python
df["Age"] = df["Age"].astype("Int64")
```

`Int64` can contain both integers and missing values:

```text
20
25
<NA>
35
```

### Important

`int64` and `Int64` are different:

| dtype   | Can store missing values? |
| ------- | ------------------------- |
| `int64` | ❌ No                      |
| `Int64` | ✅ Yes                     |

---

## Invalid Numeric Values

Sometimes a column contains values that cannot be converted to integers.

Example:

```text
Age
---
20
25
unknown
30
```

Use:

```python
df["Age"] = pd.to_numeric(
    df["Age"],
    errors="coerce"
)
```

`errors="coerce"` converts values that cannot be converted into numbers into `NaN`.

Result:

```text
20
25
NaN
30
```

The invalid value `"unknown"` becomes `NaN`.

The missing value can then be handled using an appropriate missing-value technique.

---

## Integer vs Float

### Integer

```text
10
20
30
```

Used for whole-number values.

### Float

```text
10.5
20.75
30.25
```

Used when decimal values are required.

| Type    | Example |
| ------- | ------- |
| Integer | `10`    |
| Float   | `10.5`  |

Choose the data type based on the **actual nature of the data**.

---

## Numeric-Looking Data Is Not Always Numerical

A column can contain digits but still **not be a numerical quantity**.

Example:

```text
Phone_Number
------------
9876543210
8765432109
```

Phone numbers should usually be stored as **strings**, not integers, because they are identifiers rather than quantities.

Other examples:

* Customer ID
* Account number
* Pincode
* Phone number
* Aadhaar number
* Product code

For example:

```text
Pincode
-------
600001
600002
600003
```

These are identifiers/labels, so mathematical operations such as calculating their average do not make sense.

---

## Integer Decision Rules

Before converting a column to integer, ask:

1. Is the column actually numerical?
2. Does it contain only whole-number values?
3. Does it contain decimal values?
4. Is it an identifier rather than a quantity?
5. Does it contain missing values?
6. Does it contain invalid/non-numeric values?
7. Do the values need cleaning before conversion?

---

## Key Idea

> **Use Integer for genuine whole-number quantities.**

Do not force a column to integer just because its values **look like numbers**.

Always consider the **meaning of the column** before choosing its data type.

```
```


`````python
data = {
    "Age": ["20", "25", "30", "35"],
    "Children": ["0", "2", "1", "3"],
    "Phone": ["0987654321", "9876543210", "8765432109", "7654321098"],
    "Score": ["85.5", "90.0", "78.5", "92.5"],
    "Experience": ["2", "5", "unknown", "8"]
}

df = pd.DataFrame(data)
df
`````

`````python
# Score have decimal values so we can use float
df['Score'] = df['Score'].astype('float') 
`````

`````python
# Experience we have to use integer since it can be used in future
# something like comparing experience with salary or experience with Score etc..
# There is an unknown value we cannot fill na  since his score is low doesnt fit mean and median
# so we will left behid
df['Experience'] = pd.to_numeric(df['Experience'],errors = 'coerce').astype('Int64')
`````

# FLOAT

## What is a Float?

A **float** is a numerical value that can contain a **decimal or fractional part**.

Examples:

```text
10.5
25.75
3.14
0.5
99.99
````

---

## Float in Pandas

A common Pandas float dtype is:

```python
float64
```

---

## When to Use Float?

Use **Float** when the actual meaning of the data allows decimal values.

Common examples:

* Height
* Weight
* Temperature
* Percentage
* Price
* Distance
* Average
* Measurements

Example:

```text
Height
------
170.5
165.2
180.7
```

---

## Convert String to Float

For clean numeric strings:

```python
df["Height"] = df["Height"].astype(float)
```

Example:

```text
"170.5" → 170.5
"165.2" → 165.2
```

---

## Handling Invalid Values

Example:

```text
170.5
165.2
unknown
180.7
```

Use:

```python
df["Height"] = pd.to_numeric(
    df["Height"],
    errors="coerce"
)
```

`errors="coerce"` converts values that cannot be converted into numbers into `NaN`.

Result:

```text
170.5
165.2
NaN
180.7
```

---

## Float and Missing Values

A `float64` column can contain `NaN`.

Example:

```text
20.0
25.0
NaN
35.0
```

An `int64` column normally cannot contain `NaN`.

Therefore, a column containing integers and missing values may automatically become `float64`.

Example:

```text
Original:
20
25
NaN
35

Pandas:
20.0
25.0
NaN
35.0
```

---

## Integer vs Float

### Integer

```text
10
20
30
```

### Float

```text
10.5
20.25
30.75
```

`40.0` is also a **float** because it contains a decimal representation.

---

## Don't Convert Float to Integer Blindly

Example:

```text
30500.75
40500.50
```

Converting to integer:

```python
df["Salary"] = df["Salary"].astype(int)
```

results in:

```text
30500
40500
```

The decimal portion is lost.

Therefore:

> Only convert Float to Integer when the decimal portion is not meaningful.

---

## Percentage Example

Values such as:

```text
75%
82%
91%
```

are strings because `%` is included.

Remove `%` first:

```python
df["Completion"] = (
    df["Completion"]
    .str.replace("%", "", regex=False)
)

df["Completion"] = pd.to_numeric(
    df["Completion"],
    errors="coerce"
)
```

Now the values become numerical:

```text
75
82
91
```

They can then be treated as numerical values for calculations.

---

## Decision Rule

Use **Float** when a numerical quantity can meaningfully contain decimal values.

| Column             | Typical Type           |
| ------------------ | ---------------------- |
| Height             | Float                  |
| Weight             | Float                  |
| Temperature        | Float                  |
| Price              | Float                  |
| Percentage         | Float                  |
| Distance           | Float                  |
| Average            | Float                  |
| Age                | Usually Integer        |
| Number of Children | Integer                |
| Phone Number       | String                 |
| Customer ID        | Depends on its purpose |

---

## Key Idea

> **Use Float when decimal precision is meaningful.**

Do not choose Float simply because the values look numerical.

Choose the data type based on the **meaning of the column** and whether **decimal values are meaningful**.

```
```

# DATETIME

## What is Datetime?

Datetime represents a **date and/or time**.

Examples:

```text
2026-08-12
2026-08-12 10:30:00
12/08/2026
August 12, 2026
````

---

## Incorrect Datatype

Dates may initially be stored as strings:

```text
"2026-01-10"
"2026-02-15"
```

Pandas may identify the column as:

```text
object
```

This means the dates are being treated as **text**, not as actual datetime values.

---

## Correct Datatype

Pandas uses datetime dtypes such as:

```text
datetime64[ns]
```

or:

```text
datetime64[us]
```

The exact resolution may vary depending on the data and Pandas version.

---

## Convert String to Datetime

Use `pd.to_datetime()`:

```python
df["Date"] = pd.to_datetime(df["Date"])
```

---

## Invalid Dates

Example:

```text
2026-01-15
unknown
2026-03-20
```

Use:

```python
df["Date"] = pd.to_datetime(
    df["Date"],
    errors="coerce"
)
```

`errors="coerce"` converts values that cannot be interpreted as valid dates into:

```text
NaT
```

### NaT

**NaT = Not a Time**

It represents a missing or invalid datetime value.

---

## NaN vs NaT

| Value | Meaning                  | Commonly Used For |
| ----- | ------------------------ | ----------------- |
| `NaN` | Missing numerical value  | Numeric data      |
| `NaT` | Missing/invalid datetime | Date/time data    |

---

## Date Formats

Common date formats include:

```text
DD/MM/YYYY
MM/DD/YYYY
YYYY-MM-DD
DD-MM-YYYY
```

When the format is known, it can be specified explicitly:

```python
df["Date"] = pd.to_datetime(
    df["Date"],
    format="%d/%m/%Y"
)
```

### Important Format Codes

| Code | Meaning      |
| ---- | ------------ |
| `%Y` | 4-digit year |
| `%y` | 2-digit year |
| `%m` | Month        |
| `%d` | Day          |
| `%H` | Hour         |
| `%M` | Minute       |
| `%S` | Second       |

Example:

```text
"12/08/2026"
```

with:

```python
format="%d/%m/%Y"
```

means:

**12 August 2026**

---

## Extract Date Components

After converting a column to datetime, use `.dt` to extract components.

### Year

```python
df["Date"].dt.year
```

### Month

```python
df["Date"].dt.month
```

### Day

```python
df["Date"].dt.day
```

### Day Name

```python
df["Date"].dt.day_name()
```

### Month Name

```python
df["Date"].dt.month_name()
```

---

## Filtering Dates

To filter rows based on a date:

```python
df[df["Date"] >= "2026-01-01"]
```

### Multiple Conditions

```python
df[
    (df["Date"] >= "2026-01-01") &
    (df["Date"] <= "2026-06-30")
]
```

This returns rows between **January 1, 2026 and June 30, 2026**.

---

## Sorting Dates

### Ascending

```python
df.sort_values("Date")
```

### Descending

```python
df.sort_values(
    "Date",
    ascending=False
)
```

Sorting works correctly when the column is stored as a proper datetime type.

---

## Date Arithmetic

Datetime values can be used in calculations.

Example:

```python
df["Days_Worked"] = (
    pd.Timestamp("2026-08-12")
    - df["Joining_Date"]
).dt.days
```

This calculates the number of days between the joining date and `2026-08-12`.

---

## String vs Datetime

### String

```text
"2026-08-12"
```

→ Treated as **text**.

### Datetime

```text
2026-08-12
```

→ Pandas understands it as a **date**.

Converting dates to datetime allows:

* Date sorting
* Date filtering
* Date arithmetic
* Extracting year/month/day
* Time-based feature engineering
* Time-series analysis

---

## Important Rule

**Never blindly assume the date format.**

For example:

```text
01/02/2026
```

could mean:

```text
1 February 2026
```

or:

```text
January 2, 2026
```

depending on the dataset's date convention.

Always understand the **date format and convention** before conversion.

---

## Key Idea

# CATEGORY

## What is Category?

A **categorical variable** contains a limited number of distinct groups or labels.

Examples:

- Gender
- City
- Department
- Size
- Product Type
- Blood Group

---

## Category in Pandas

Pandas provides a special dtype:

```python
category
````

Example:

```python
df["City"] = df["City"].astype("category")
```

Check the dtype:

```python
df["City"].dtype
```

---

## String vs Category

### String

Generally suitable for **free-form text**:

* Name
* Address
* Email
* Comments

### Category

Suitable for **limited, repeated groups**:

* City
* Gender
* Department
* Size
* Product Type
* Blood Group

---

## Why Use Category?

The `category` dtype can reduce **memory usage** when a column contains many repeated values but only a small number of unique categories.

Example:

```text
City
------
Chennai
Chennai
Madurai
Chennai
Trichy
Madurai
```

There are many repeated values but only a few unique categories.

Instead of storing the full text repeatedly, Pandas can internally represent the data using categories and codes.

---

## Convert String to Category

```python
df["City"] = df["City"].astype("category")
```

---

## View Categories

```python
df["City"].cat.categories
```

This returns the unique categories currently stored in the column.

---

## Count Unique Categories

Using `nunique()`:

```python
df["City"].nunique()
```

Or using the category list:

```python
df["City"].cat.categories.size
```

---

## Adding Categories

A new category must be added before assigning it to a categorical column.

```python
df["City"] = df["City"].cat.add_categories(
    ["Coimbatore"]
)
```

Then the new category can be assigned.

---

## Removing Unused Categories

After filtering a dataset, some categories may no longer have any rows but can still remain in the category list.

Remove unused categories using:

```python
df["City"] = (
    df["City"]
    .cat.remove_unused_categories()
)
```

---

## Standardize Before Converting to Category

Example:

```text
Chennai
chennai
CHENNAI
Madras
```

These represent the same city but are written differently.

First clean and standardize:

```python
df["City"] = (
    df["City"]
    .str.strip()
    .str.lower()
    .replace({
        "madras": "chennai"
    })
)
```

Then convert to category:

```python
df["City"] = df["City"].astype("category")
```

### Correct Workflow

```text
Clean
  ↓
Standardize
  ↓
Convert to Category
```

---

## Nominal Categories

**Nominal categories** are categories with **no natural or meaningful order**.

Example:

```text
City:
Chennai
Madurai
Trichy
```

There is no meaningful order such as:

```text
Chennai < Madurai < Trichy
```

Therefore, these are nominal categories.

---

## Ordinal Categories

**Ordinal categories** have a **meaningful order or ranking**.

Example:

```text
Small
Medium
Large
```

Here:

```text
Small < Medium < Large
```

Create an ordered category using:

```python
df["Size"] = pd.Categorical(
    df["Size"],
    categories=["Small", "Medium", "Large"],
    ordered=True
)
```

Now Pandas understands the order:

```text
Small < Medium < Large
```

This is useful for sorting and comparisons.

---

## Nominal vs Ordinal

| Type        | Meaning          | Example                   |
| ----------- | ---------------- | ------------------------- |
| **Nominal** | No natural order | City, Gender, Blood Group |
| **Ordinal** | Meaningful order | Small, Medium, Large      |

---

## Important

`category` does **not automatically mean numerical**.

Do not automatically assign:

```text
Small = 1
Medium = 2
Large = 3
```

unless there is a specific reason to represent the ordinal categories numerically.

For machine learning, categorical variables may need to be encoded depending on the algorithm and the type of category.

---

## Key Idea

Use **Category** when a column contains a **limited set of repeated labels representing groups**.

```text
String   → Free-form text
Category → Limited, repeated groups
```

### Remember

**Nominal → Categories without order**

**Ordinal → Categories with meaningful order**

`````python
df['City'] = df['City'].str.title().str.strip()
df['Department'] = df['Department'].str.title().str.strip()
df['Size'] = df['Size'].str.title().str.strip()
`````

`````python
df['City'] = df['City'].astype('category')
df['Department'] = df['Department'].astype('category')
df['Size'] = df['Size'].astype('category')
`````
