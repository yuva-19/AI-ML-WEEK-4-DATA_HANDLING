
# STRING CLEANING



String cleaning is the process of **standardizing text data and removing unwanted characters, spaces, or formatting**.

Main techniques:

1. Regular Expressions
2. Whitespace Removal
3. Case Normalization
4. Special Character Removal
5. String Replacement and Extraction
6. Cleaning Before Datatype Conversion

---

# 1. REGULAR EXPRESSIONS (REGEX)

**Regex (Regular Expression)** is used for **pattern matching and text manipulation**.

Common Regex patterns:

| Pattern | Meaning |
|---|---|
| `\d` | Digit |
| `\D` | Non-digit |
| `\w` | Word character |
| `\W` | Non-word character |
| `\s` | Whitespace |
| `\S` | Non-whitespace |
| `.` | Any character |
| `+` | One or more |
| `*` | Zero or more |
| `?` | Zero or one |
| `^` | Start of string |
| `$` | End of string |
| `[]` | Character set |
| `()` | Capture group |

Example:

```python
r"\d+"
````

This matches **one or more digits**.

Example:

```text
"Age: 25"
```

`r"\d+"` matches:

```text
25
```

---

# 2. WHITESPACE REMOVAL

## Remove Leading and Trailing Spaces

Use `.str.strip()`:

```python
df["Name"] = df["Name"].str.strip()
```

Example:

```text
"  Arun Kumar  "
```

↓

```text
"Arun Kumar"
```

---

## Remove Leading Spaces

Use `.str.lstrip()`:

```python
df["Name"] = df["Name"].str.lstrip()
```

---

## Remove Trailing Spaces

Use `.str.rstrip()`:

```python
df["Name"] = df["Name"].str.rstrip()
```

---

## Remove Multiple Spaces Between Words

Use Regex:

```python
df["Name"] = df["Name"].str.replace(
    r"\s+",
    " ",
    regex=True
)
```

Example:

```text
"Arun    Kumar"
```

↓

```text
"Arun Kumar"
```

### Complete Whitespace Cleaning

```python
df["Name"] = (
    df["Name"]
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
)
```

---

# 3. CASE NORMALIZATION

Case normalization makes text follow a consistent capitalization format.

## Lowercase

```python
df["City"] = df["City"].str.lower()
```

Example:

```text
"CHENNAI"
```

↓

```text
"chennai"
```

---

## Uppercase

```python
df["City"] = df["City"].str.upper()
```

Example:

```text
"chennai"
```

↓

```text
"CHENNAI"
```

---

## Capitalize

```python
df["Name"] = df["Name"].str.capitalize()
```

Example:

```text
"ARUN KUMAR"
```

↓

```text
"Arun kumar"
```

---

## Title Case

```python
df["Name"] = df["Name"].str.title()
```

Example:

```text
"arun kumar"
```

↓

```text
"Arun Kumar"
```

### When to Use

For categorical values, lowercase is often useful for standardization:

```text
Chennai
CHENNAI
chennai
```

↓

```text
chennai
chennai
chennai
```

For names, title case is often more readable.

---

# 4. SPECIAL CHARACTER REMOVAL

## Remove Everything Except Letters and Whitespace

```python
df["Name"] = df["Name"].str.replace(
    r"[^a-zA-Z\s]",
    "",
    regex=True
)
```

Example:

```text
"Arun@123 Kumar!"
```

↓

```text
"Arun Kumar"
```

---

## Remove Everything Except Digits

Use `\D`:

```python
df["Phone"] = df["Phone"].str.replace(
    r"\D",
    "",
    regex=True
)
```

Example:

```text
"+91-98765-43210"
```

↓

```text
"919876543210"
```

This is useful when phone numbers contain spaces, hyphens, brackets, or other formatting characters.

---

# 5. CURRENCY CLEANING

Example:

```text
₹30,000
₹45,000
₹50,000
```

Remove the currency symbol and commas:

```python
df["Price"] = df["Price"].str.replace(
    r"[₹,]",
    "",
    regex=True
)
```

Result:

```text
30000
45000
50000
```

Then convert to a numerical datatype:

```python
df["Price"] = pd.to_numeric(
    df["Price"],
    errors="coerce"
)
```

Now the values can be used for numerical calculations.

---

# 6. STRING REPLACE

`.str.replace()` can replace specific text.

Example:

```python
df["City"] = df["City"].str.replace(
    "Madras",
    "Chennai",
    regex=False
)
```

This changes:

```text
Madras
```

to:

```text
Chennai
```

It can also be used with Regex patterns:

```python
df["Name"] = df["Name"].str.replace(
    r"\d+",
    "",
    regex=True
)
```

This removes digits from the string.

---

# 7. REGEX `replace()` VS `extract()`

## `replace()`

Used to **remove or replace unwanted text**.

Example:

```python
df["Name"] = df["Name"].str.replace(
    r"\d+",
    "",
    regex=True
)
```

---

## `extract()`

Used to **extract specific information from a string using a pattern**.

Example:

```python
df["Domain"] = df["Email"].str.extract(
    r"@(.+)"
)
```

For:

```text
user@gmail.com
```

Result:

```text
gmail.com
```

---

# 8. COMPLETE STRING CLEANING PIPELINE

Raw value:

```text
"  ARUN!!!    KUMAR  "
```

Clean it using:

```python
df["Name"] = (
    df["Name"]
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
    .str.replace(r"[^a-zA-Z\s]", "", regex=True)
    .str.title()
)
```

Result:

```text
"Arun Kumar"
```

### Cleaning Steps

```text
Raw String
    ↓
Remove leading/trailing spaces
    ↓
Remove extra spaces
    ↓
Remove unwanted characters
    ↓
Standardize case
    ↓
Clean String
```

---

# 9. STRING CLEANING BEFORE DATATYPE CONVERSION

String cleaning is often required before converting a column to a numerical or datetime datatype.

Example:

```text
" ₹30,000 "
```

First clean the string:

```text
"30000"
```

Then convert the datatype:

```python
df["Price"] = pd.to_numeric(
    df["Price"],
    errors="coerce"
)
```

Result:

```text
30000.0
```

### General Workflow

```text
Raw Data
   ↓
Clean
   ↓
Standardize
   ↓
Validate
   ↓
Convert Datatype
```

---

# IMPORTANT RULES

* Use `.str.strip()` to remove leading and trailing whitespace.
* Use `.str.replace()` for replacing or removing unwanted text.
* Use Regex when the unwanted text follows a pattern.
* Use `.str.lower()` or `.str.upper()` for consistent categorical values.
* Use `.str.title()` when title case is appropriate for text such as names.
* Clean currency symbols and separators before numerical conversion.
* Use `errors="coerce"` when invalid values should become missing values.
* Do not remove characters blindly; determine whether they contain useful information.
* Always consider the meaning of the data before cleaning.

---
