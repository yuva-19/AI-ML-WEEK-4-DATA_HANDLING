# DATE & TIME PROCESSING



Date and time processing is used to **clean, convert, manipulate, compare, and extract useful information from datetime data**.

Main topics:

1. Parsing
2. Time Zones
3. Date Arithmetic
4. Feature Extraction

---

# 1. PARSING

**Parsing** means converting date/time strings into proper Pandas datetime objects.

Example:

```text
"15/01/2024"
````

↓

```text
datetime64
```

### Pandas

```python
df["Joining_Date"] = pd.to_datetime(
    df["Joining_Date"],
    format="%d/%m/%Y"
)
```

Check the datatype:

```python
df.dtypes
```

### Handling Invalid Dates

Invalid values can be converted to `NaT` using `errors="coerce"`:

```python
df["Joining_Date"] = pd.to_datetime(
    df["Joining_Date"],
    format="%d/%m/%Y",
    errors="coerce"
)
```

`NaT` = **Not a Time**

It represents a missing or invalid datetime value.

# 2. TIME ZONES

A **time zone** identifies the local time associated with a datetime.

### Naive Datetime

```text
2026-08-12 10:00:00
```

No timezone information is attached.

### Timezone-Aware Datetime

```text
2026-08-12 10:00:00+05:30
```

The timezone information is included.

---

## `tz_localize()`

Used to **attach a timezone** to a timezone-naive datetime.

```python
df["Date"] = df["Date"].dt.tz_localize(
    "Asia/Kolkata"
)
```

Example:

```text
2026-08-12 10:00:00
```

↓

```text
2026-08-12 10:00:00+05:30
```

---

## `tz_convert()`

Used to **convert a timezone-aware datetime to another timezone**.

```python
df["Date"] = df["Date"].dt.tz_convert(
    "America/New_York"
)
```

### Remember

```text
tz_localize → Attach a timezone
tz_convert  → Convert to another timezone
```

`tz_convert()` requires the datetime to already be timezone-aware.

# 3. DATE ARITHMETIC

Once dates are converted to datetime, they can be **added, subtracted, and compared**.

## Difference Between Dates

```python
df["End_Date"] - df["Start_Date"]
```

This returns a `Timedelta`.

### Difference in Days

```python
(df["End_Date"] - df["Start_Date"]).dt.days
```

Example:

```text
Start_Date → 2026-01-01
End_Date   → 2026-01-10
```

Result:

```text
9 days
```

---

## Add Days

Use `pd.Timedelta()`:

```python
df["Date"] + pd.Timedelta(days=30)
```

---

## Add Months

Use `pd.DateOffset()`:

```python
df["Date"] + pd.DateOffset(months=1)
```

---

## Add Years

```python
df["Date"] + pd.DateOffset(years=1)
```

---

## Subtract Dates

```python
df["End_Date"] - df["Start_Date"]
```

The result is a `Timedelta` representing the difference between the two dates.

# 4. FEATURE EXTRACTION

Datetime columns contain multiple useful components.

Use the **`.dt` accessor** to extract them.

### Year

```python
df["Year"] = df["Date"].dt.year
```

### Month

```python
df["Month"] = df["Date"].dt.month
```

### Day

```python
df["Day"] = df["Date"].dt.day
```

### Day of Week

```python
df["DayOfWeek"] = df["Date"].dt.dayofweek
```

`dayofweek` uses:

```text
Monday    → 0
Tuesday   → 1
Wednesday → 2
Thursday  → 3
Friday    → 4
Saturday  → 5
Sunday    → 6
```

### Day Name

```python
df["DayName"] = df["Date"].dt.day_name()
```

Example:

```text
Monday
Tuesday
Wednesday
```

### Month Name

```python
df["MonthName"] = df["Date"].dt.month_name()
```

### Hour

```python
df["Hour"] = df["Date"].dt.hour
```

### Minute

```python
df["Minute"] = df["Date"].dt.minute
```

### Second

```python
df["Second"] = df["Date"].dt.second
```

### Quarter

```python
df["Quarter"] = df["Date"].dt.quarter
```

Quarter values:

```text
1 → January to March
2 → April to June
3 → July to September
4 → October to December
```

# DATE & TIME PROCESSING WORKFLOW

```text
Raw Date/Time Strings
        ↓
      Parse
        ↓
Convert to Datetime
        ↓
Handle Invalid Dates
        ↓
Handle Time Zones if Required
        ↓
Perform Date Arithmetic
        ↓
Extract Useful Features
        ↓
Use for Analysis / Machine Learning
```

# MACHINE LEARNING USE

Datetime itself is often converted into useful numerical or categorical features for machine learning.

Example:

```text
Datetime
   ↓
Year
Month
Day
DayOfWeek
Hour
Quarter
```

These features can help a machine learning model identify **time-based patterns**.

Examples:

* Sales by month
* Sales by day of week
* Website traffic by hour
* Seasonal demand
* Customer activity by time
* Machine failures by time
* Electricity consumption by hour

`````python
import pandas as pd
data = {
    "Employee": [
        "Arun", "Priya", "Kumar",
        "Meena", "Ravi"
    ],

    "Joining_Date": [
        "15/01/2024",
        "20/03/2024",
        "10/06/2025",
        "12/08/2025",
        "01/02/2026"
    ],

    "Last_Promotion": [
        "15/01/2025",
        "20/09/2025",
        "10/06/2026",
        "12/08/2026",
        "01/08/2026"
    ]
}

df = pd.DataFrame(data)
`````

`````python
df['Joining_Date'] = pd.to_datetime(df['Joining_Date'],format = '%d/%m/%Y')
`````

`````python
df['Last_Promotion'] = pd.to_datetime(df['Last_Promotion'],format = '%d/%m/%Y')
`````

`````python
df["Days_To_Promotion"] = (
    df["Last_Promotion"] -
    df["Joining_Date"]
).dt.days
`````

`````python
# Joining_Year
# Joining_Month
# Joining_Day
# Joining_DayName

df['Joining_Year'] = df['Joining_Date'].dt.year
df['Joining_Month'] = df['Joining_Date'].dt.month
df['Joining_Day'] = df['Joining_Date'].dt.day
df['Joining_DayName'] = df['Joining_Date'].dt.day_name()
`````

`````python
# Promotion_Year
# Promotion_Month
# Promotion_Quarter

df['Promotion_Year'] = df['Last_Promotion'].dt.year
df['Promotion_Month'] = df['Last_Promotion'].dt.month
df['Promotion_Quarter'] = df['Last_Promotion'].dt.quarter
`````

`````python
df["Joining_Date_IST"] = df["Joining_Date"].dt.tz_localize(
    "Asia/Kolkata"
)
`````

`````python
df["Joining_Date_NY"] = df["Joining_Date_IST"].dt.tz_convert(
    "America/New_York"
)
`````
