# Numpy Bascis

```python
import numpy as np
print(np.__version__)
```

```python
# ndarray = N-dimensional array.
# Stores only one dtype.
# Has attributes like shape, ndim, size, and dtype.
# Memory efficient compared to Python lists.
# Supports broadcasting and vectorization.
# Operations are implemented in optimized C code.
# Fixed-size container.
# Most ML datasets become NumPy arrays before training.
```

```python
arr=np.array([1,2,3,4])
print(arr)
print(type(arr))
print(arr.shape)
```

```python
# shape == Returns the size of each dimension of an array.

arr1=np.array([[1,2,3,4],[2,3,6,8]])
print(arr1.shape)
```

```python
# ndim	Returns the number of axes (dimensions) in an array.
# 1D Array	One axis → ndim = 1
# 2D Array	Rows & columns → ndim = 2
# 3D Array	Depth, rows & columns → ndim = 3
arr2=np.array([[1,2,3],[1,5,7],[3,6,9]])
print(arr2.ndim)

image=np.zeros((224,224,3))
print(image.ndim)
```

```python
# size==Returns the total number of elements in an array.
print(image.size)
```

```python
# dtype	Returns the data type of the array.
# dtype=	Specifies the data type during array creation.
# astype()	Converts an array to a new data type.

print(image.dtype)
arr=np.array([1,2,3],dtype=np.float32)
print(arr.dtype)
arr=arr.astype(np.float64)
print(arr.dtype)
```

```python
# Memory Layout	Defines how multi-dimensional arrays are stored in memory.
# C-order	Row-major order (rows stored first). NumPy default.
# Fortran-order	Column-major order (columns stored first).
# Performance	Accessing data in its storage order is faster due to cache locality.
# Importance	Useful when optimizing performance or interfacing with libraries like BLAS/LAPACK.

arr_c = np.array([[1, 2],
                  [3, 4]], order='C')

arr_f = np.array([[1, 2],
                  [3, 4]], order='F')

print(arr_c)
print(arr_f)
```

```python
# strides Number of bytes NumPy skips in memory to move to the next element along each axis.
print(arr_c.strides)
# 16 indicates the row
# 8 indicates the coloumn
```

# Broadcasting

```python
# Broadcasting is NumPy's mechanism for performing element-wise operations between arrays of 
# different shapes by automatically expanding the smaller array when possible.

a = np.array([1,2,3])
b = 10 #b treated as [10 10 10]
print(a + b)
```

```python
# Broadcasting Rules
# ✅ They are equal
# ✅ One of them is 1
# ❌ Otherwise → Broadcasting Error
```

```python
# a=(4,3)
# b=(3,)now its reshaped as (,3)--->(4,3)
```

```python
# a=(4,3)
# b=(4,1)--->(4,3)
```

```python
# a=(3,4,2)
# b=(4,2)--->(3,4,2)
```

```python
# a=(3,4)
# b=(3,2)-->broadcasting error because there is no 1 and its not same shape so it generates error
```

```python
# Performance Benchmarking
# Use the time or timeit module to compare performance.

# python loop

import time
import numpy as np

a = np.arange(1_000_000)

start = time.time()

result = []

for x in a:
    result.append(x * 2)

print(time.time() - start)
```

```python
# Numpy Vectorization
start = time.time()

result = a * 2

print(time.time() - start)
```

```python
# Better Benchmarking with timeit

import timeit
timeit.timeit("a*2", globals=globals(), number=100)
```

# NumPy Boolean Masking

```python
a = np.array([10,20,30,40,50])
mask = a > 25
print(mask)
print(a[mask])
```

```python
# Filtering Rows
marks=np.array([10,20,40,60,70,80,90])
passed=marks[marks>=35]
print(passed)
print()

students = np.array([[1,85],
                     [2,40],
                     [3,95]])
print(students[students[:,1]>=50])
```

```python
# Conditional Replacement
a=np.array([-10,20,30,-40,50,60,-70,80,-90])
a[a < 0]=0 #remove negative values
print(a)

```

```python
# np.where(condition, value_if_true, value_if_false)

a = np.array([10,20,30,40,-50,-25])
b = np.where(a<0 ,a*(-1), a) #changes the negative to positive
print(b)
```

```python
# Data Selection

a = np.array([5,10,15,20])
print(a[a > 10])
print(a[(a > 10) & (a < 20)]) #AND operation
print(a[(a < 10) | (a > 15)])# OR operation
print(a[~(a > 10)])#NOT operation
```

# NumPy Random Module

```python
# Random Float (0 to 1)
rng = np.random.default_rng()
rng.random()
```

```python
# Multiple Random Floats
rng.random(5)
```

```python
# Random Integer
print(rng.integers(1, 10))
```

```python
# 2D Random Array
rng.random((2,3))
```

```python
# Random Seed
# Purpose
# A seed ensures the same sequence of random numbers is generated every time.
rng = np.random.default_rng(42)
print(rng.random(3))
```

```python
# Random Distributions
# Uniform Distribution
rng.uniform(10, 20, 5)
```

```python
# Normal (Gaussian) Distribution

# loc → Mean
# scale → Standard deviation
# size → Number of samples

rng.normal(loc=0, scale=1, size=5)
```

```python
# Binomial Distribution
rng.binomial(n=10, p=0.5, size=5)
```

```python
# Choice from Values
rng.choice([10,20,30], size=5)
```

```python
# Shuffling 
# shuffle()
# Shuffles in-place.
# original gets changed
a = np.array([1,2,3,4,5])
rng.shuffle(a)
print(a)
```

```python
# permutation()
# Returns a shuffled copy.

a = np.array([1,2,3,4,5])
b = rng.permutation(a)
print(b)
print(a)
```

```python
# Sampling ⭐⭐⭐⭐⭐
# Random Sample Without Replacement
# No duplicates.
rng.choice(10, size=5, replace=False)
```

```python
# Sample With Replacement
# Duplicates allowed.
rng.choice(10, size=5, replace=True)
```

```python
# Reproducibility 
# Reproducibility means getting the same results every time.
# Without seed

rng = np.random.default_rng()
print(rng.random(3))
```

```python
# With seed

rng = np.random.default_rng(42)
print(rng.random(3))
```
