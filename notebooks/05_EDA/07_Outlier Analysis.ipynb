# Outlier Analysis 🚨

> An outlier is an observation that is unusually different from the rest of the data.

**Important:** Outlier ≠ Error. It may be a genuine extreme observation.

## 1. Visual Methods

### Box Plot

```python
import seaborn as sns

sns.boxplot(x=df["Salary"])
````

Potential outliers appear beyond the whiskers.

### Histogram

```python
df["Salary"].hist()
```

Useful for spotting long tails/extreme values.

### Scatter Plot

```python
sns.scatterplot(
    x=df["Age"],
    y=df["Salary"]
)
```

Useful for spotting unusual points in relationships.

---

## 2. Statistical Methods

### IQR Method

```python
Q1 = df["Salary"].quantile(0.25)

Q3 = df["Salary"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df["Salary"] < lower) |
    (df["Salary"] > upper)
]
```

### Z-Score

```python
from scipy.stats import zscore

df["z_score"] = zscore(
    df["Salary"]
)
```

Common rule:

`|Z| > 3` → Potential outlier

---

## 3. Domain-Specific Method

Use domain knowledge to determine whether an unusual value is:

* Valid
* Data-entry error
* Measurement error
* Rare but genuine

Example:

`Age = 150` → likely invalid

`CEO Salary = ₹50 lakh` → may be valid

## What To Do With Outliers

Depending on the reason:

* Keep
* Correct
* Remove
* Cap/Winsorize
* Transform

**Never automatically delete outliers.**

## Key Workflow

**Detect → Investigate → Understand Cause → Decide → Document**

## Key Idea

**Visual → What looks unusual?**

**Statistical → What is mathematically unusual?**

**Domain-specific → Is it actually wrong or valid?**

```
```
