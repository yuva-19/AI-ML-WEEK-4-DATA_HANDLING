
# Distribution Analysis 📊

> Studies how the values of a numerical variable are distributed.

## 1. Normality

Normal distribution → symmetric, bell-shaped distribution.

Properties:
- Mean ≈ Median ≈ Mode
- Symmetric
- Bell-shaped

### Histogram

```python
df["Salary"].hist()
````

### Q-Q Plot

```python
import scipy.stats as stats
import matplotlib.pyplot as plt

stats.probplot(
    df["Salary"],
    dist="norm",
    plot=plt
)

plt.show()
```

Points close to the diagonal → approximately normal.

### Shapiro-Wilk Test

H0 → Data is normally distributed

H1 → Data is not normally distributed

```python
from scipy.stats import shapiro

stat, p = shapiro(df["Salary"])

print(p)
```

`p < 0.05` → Reject H0

`p >= 0.05` → Fail to reject H0

Use visual checks along with the test.

---

## 2. Skewness

Measures asymmetry of a distribution.

```python
df["Salary"].skew()
```

* `≈ 0` → Symmetric
* `> 0` → Right/Positive skew
* `< 0` → Left/Negative skew

Right skew → typically `Mean > Median`

Left skew → typically `Mean < Median`

---

## 3. Kurtosis

Describes tail behavior/extreme observations.

```python
df["Salary"].kurt()
```

Pandas uses **excess kurtosis**:

* `≈ 0` → Normal-like
* `> 0` → Heavier tails
* `< 0` → Lighter tails

---

## 4. Density Estimation

KDE (Kernel Density Estimation) gives a smooth estimate of the distribution.

```python
import seaborn as sns

sns.kdeplot(df["Salary"])
```

### Histogram + KDE

```python
sns.histplot(
    df["Salary"],
    kde=True
)
```

KDE helps identify:

* Center
* Spread
* Peaks
* Skewness
* Distribution shape

---

## Key Workflow

**Histogram → Box Plot → Skewness → Kurtosis → KDE → Q-Q Plot → Normality Test**

## Key Idea

Normality → Is the distribution approximately normal?

Skewness → Is it asymmetric?

Kurtosis → How heavy/light are the tails?

KDE → What is the smooth distribution shape?
