# Multivariate Analysis 📊

> Analysis of **3 or more variables simultaneously** to identify relationships and patterns.

## 1. Pairwise Relationships

Examines relationships between every pair of variables.

### Correlation Matrix

```python
df.corr(numeric_only=True)
````

→ Shows the **strength and direction of linear relationships** between numerical variables.

### Pair Plot

Visualizes pairwise relationships between numerical variables.

```python
import seaborn as sns

sns.pairplot(df)
```

* Scatter plots → relationships between variables
* Diagonal → individual distributions

---

## 2. High-Dimensional Patterns

Datasets with many variables/features are difficult to analyze directly.

### Correlation Heatmap

Useful for identifying relationships among many numerical variables.

```python
import seaborn as sns
import matplotlib.pyplot as plt

corr = df.corr(numeric_only=True)

sns.heatmap(
    corr,
    annot=True
)

plt.show()
```

### Key Idea

Multivariate analysis helps identify:

* Strong/weak relationships
* Groups of related variables
* Patterns across multiple variables
* Potentially redundant features
* Possible multicollinearity

## Important

**Correlation ≠ Causation**

A strong correlation does not prove that one variable causes another.

## Key Workflow

**Correlation Matrix → Pair Plot → Heatmap → Identify Patterns → Interpret**
