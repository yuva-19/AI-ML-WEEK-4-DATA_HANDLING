
Useful for detecting:

* Class imbalance
* Rare classes
* Dominant classes

### Regression Target

```python
df["Target"].describe()

df["Target"].hist()

df.boxplot(
    column="Target"
)
```

Check:

* Center
* Spread
* Distribution
* Skewness
* Outliers

---

## 2. Feature–Target Relationships

Analyze how each feature relates to the target.

### Numerical → Numerical

Use:

* Scatter plot
* Correlation

```python
sns.scatterplot(
    x="Feature",
    y="Target",
    data=df
)

df["Feature"].corr(
    df["Target"]
)
```

### Categorical → Numerical

Use:

* Group statistics
* Box plot

```python
df.groupby(
    "Category"
)["Target"].mean()
```

```python
sns.boxplot(
    x="Category",
    y="Target",
    data=df
)
```

### Numerical → Categorical

Compare numerical distributions across target classes.

```python
sns.boxplot(
    x="Target",
    y="Feature",
    data=df
)
```

### Categorical → Categorical

Use:

* Crosstab
* Percentages
* Chi-square

```python
pd.crosstab(
    df["Feature"],
    df["Target"]
)
```

---

## 3. Multicollinearity

> Multicollinearity occurs when predictor/features are highly related to each other.

Problems:

* Unstable regression coefficients
* Difficult coefficient interpretation
* Increased standard errors

### Correlation Matrix

```python
features = [
    "X1",
    "X2",
    "X3"
]

df[features].corr()
```

Use a heatmap:

```python
sns.heatmap(
    df[features].corr(),
    annot=True
)
```

---

## 4. Variance Inflation Factor (VIF)

> VIF measures how much the variance of a regression coefficient is inflated due to relationships among predictors.

```python
from statsmodels.stats.outliers_influence import (
    variance_inflation_factor
)
```

```python
X = df[features]

vif = pd.DataFrame()

vif["Feature"] = X.columns

vif["VIF"] = [
    variance_inflation_factor(
        X.values,
        i
    )
    for i in range(X.shape[1])
]

vif
```

### Rough Interpretation

`≈ 1` → Little/no multicollinearity

`1–5` → Usually acceptable

`5–10` → Potentially high

`>10` → Often severe

**Do not include the target when calculating VIF.**

## Key Workflow

**Analyze Target → Feature–Target Relationships → Feature Correlations → Multicollinearity → VIF → Feature Selection**

## Important

High VIF does not automatically mean a feature must be removed.

Use:

**Statistics + Domain Knowledge + Model Purpose**

```
```
