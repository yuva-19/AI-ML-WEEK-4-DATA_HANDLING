# EDA — Dataset Overview

## Shape

df.shape

Returns:
(rows, columns)

- Rows = observations/samples.
- Columns = variables/features/target

Always inspect dataset size before ML.

## Basic Inspection

- df.head()       → first 5 rows
- df.tail()       → last 5 rows
- df.columns      → column names
- df.dtypes       → datatype of each column
- df.info()       → structure, non-null counts, datatypes
- df.describe()   → numerical summary statistics

## Common Data Types

- int64 / float64  → numerical
- object / string  → text/categorical
- bool             → True/False
- datetime64       → date/time

## ML Mindset

Don't just execute commands.
Interpret the output.

Identify:
- number of observations
- number of features
- target variable
- numerical features
- categorical features
- possible data-quality issues

# Missing Values

## Detection

```python
df.isna().sum()
````

→ number of missing values per column

```python
df.isna().mean() * 100
```

→ percentage of missing values per column

## Common Handling

Numerical:

* Mean
* Median

Categorical:

* Mode
* "Unknown"

Drop:

```python
df.dropna()
df.drop(columns=["column"])
```

Fill:

```python
df["Age"].fillna(df["Age"].median())
df["City"].fillna("Unknown")
```

## Important

Do NOT blindly drop/impute.

Consider:

* amount missing
* why values are missing
* feature importance
* dataset size
* business/domain meaning

## Missing Indicator

```python
df["feature_missing"] = df["feature"].isna().astype(int)
```

Sometimes missingness itself contains useful information.

## ML Rule

Split data BEFORE learning imputation statistics.

Learn preprocessing parameters from training data only.

Never allow test/validation information to influence preprocessing.

```
```


# Cardinality

Cardinality = number of unique values in a column.

```python
df["City"].nunique()
````

→ number of unique values

```python
df["City"].unique()
```

→ actual unique values

```python
df["City"].value_counts()
```

→ frequency of each category

```python
df.nunique()
```

→ cardinality of all columns

## ML Considerations

Low cardinality:
→ usually easier to encode

High cardinality:
→ investigate before encoding
→ can increase feature space
→ can cause overfitting
→ may increase memory/training cost

High cardinality ≠ automatically drop.
Consider the meaning of the feature.

# Class Distribution

```python
df["target"].value_counts()
```

→ number of samples in each class

```python
df["target"].value_counts(normalize=True) * 100
```

→ percentage of each class

Balanced:
→ classes have roughly similar representation

Imbalanced:
→ one class dominates

## Important

Always inspect class distribution before classification.

Majority-class baseline:
→ accuracy obtained by always predicting the most frequent class.

For imbalanced datasets:
→ accuracy alone can be misleading.

Useful metrics may include:
Precision, Recall, F1, ROC-AUC, PR-AUC.
