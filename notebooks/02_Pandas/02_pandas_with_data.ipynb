```python
us_grp=country_grp.get_group('United States')
```

```python
us_grp['SocialMedia'].value_counts()
```

```python
# or we can
filt = (df['Country'] == 'India')
df.loc[filt]['SocialMedia'].value_counts()
```

```python
# we cannot change values one by one in filt to see everything so 
# we can use that grouped

country_grp['SocialMedia'].value_counts().head(50)
```

```python
# To get a specific Country
country_grp['SocialMedia'].value_counts(normalize=True).loc['Russian Federation']
```

```python
# to get a median salary
country_grp['ConvertedComp'].median().head(50)
```

```python
country_grp['ConvertedComp'].median().loc['India']
```

```python
# agg() function used to get things we want at once like mean median.. in column wise data
country_grp['ConvertedComp'].agg(['median','mean']).head(50)
```

```python
country_grp['ConvertedComp'].agg(['median','mean']).loc[['India','China','United States']]
```

```python
# Series → .str works directly-->df['LanguageWorkedWith']
# GroupBy → use .apply() to operate on each group

# how many of them worked with specific langugage using apply function
country_grp['LanguageWorkedWith'].apply(lambda x: x.str.contains('Python').sum())
```

```python
country_grp['LanguageWorkedWith'].apply(lambda x: x.str.contains('Python').sum()).loc[['India','United States','China','Pakistan']]
```

```python
# To know percentage of people using python from each country
country_grp['LanguageWorkedWith'].apply(lambda x: x.str.contains('Python',na=False).mean()*100)
```

```python
# another method
country_grp['LanguageWorkedWith'].apply(lambda x: x.str.contains('Python').sum()/len(x) * 100)
```

```python
# another methods
country_respondents = df['Country'].value_counts()
```

```python
knows_python = country_grp['LanguageWorkedWith'].apply(lambda x: x.str.contains('Python').sum())
```

```python
python_df = pd.concat([country_respondents,knows_python],axis='columns',sort=False)
```

```python
python_df
```

```python
python_df.rename(columns= {'count':'Respondents','LanguageWorkedWith': 'Knowspython'},inplace=True)
```

```python
python_df['PercentknowsPython'] =(python_df['Knowspython']/python_df['Respondents']) * 100
```

```python
python_df
```

```python
python_df.sort_values(by='PercentknowsPython',ascending=False,inplace=True)
```

 # Working with Dates and Time Series Data

```python
import pandas as pd
```

```python
df=pd.read_csv('data/ETH_1h.csv')
```

```python
df.shape
```

```python
df.head(5)
```

```python
df.drop(columns=['Unix Timestamp'],inplace=True)
```

```python
df['Date'] = pd.to_datetime(df['Date'],format = '%Y-%m-%d %H:%M:%S')
# since we have format already like that we dont need to run
# if we have something like 8pm and dd-mm-yy

# we have to convert that

#  https://docs.python.org/3/library/datetime.html#strftime-and-strptime-behavior
# use the above like for conversion process
```

```python
df.loc[0,'Date'].day_name()
```

```python
# To get all the Day name we have to use .dt
df['Date'].dt.day_name()
```

```python
df['Date'].min()
```

```python
df['Date'].max()
```

```python
df['Date'].max() - df['Date'].min()
```

```python
# if we wanna data from 2020
filt = (df['Date'] >= '2020' )
```

```python
df.loc[filt]
```

```python
filt = (df['Date'] > '2019') & (df['Date'] <= '2020')
```

```python
df.loc[filt]
```

```python
# to use complete date format
filt = (df['Date'] >= ('2019-01-19')) & (df['Date'] <= ('2020-01-19'))
```

```python
# To get a mean value of Close for a particular timeperiod
df.loc[filt,['High','Low','Close']].mean()
```

```python
# if u wanna do for a particular date
# filt = (df['Date'] == ('2020-04-15')) # this only checks 2020-04-15 00:00:00 so to get everything
filt = (
    (df['Date'] >= '2019-01-19') &
    (df['Date'] < '2019-01-20')
)
df.loc[filt]
```

```python
# To get mean values for particular day
df.loc[filt,['High','Low','Close']].mean()
```

```python
df.loc[filt,['High','Low','Close']].max()
```

```python
df.loc[filt,['High','Low','Close']].min()
```

```python
df.loc[filt,['High','Low','Close']].median()
```

```python
# use the above link to know how to use it like if u wanna weekly daily monthly yearly etc..
df = df.set_index('Date')
```

```python
# if u just wanna get high value from each day instead all 
# you can use sample 
# https://pandas.pydata.org/pandas-docs/stable/user_guide/timeseries.html#dateoffset-objects
highs = df['High'].resample('D').max()
```

```python
%matplotlib inline
```

```python
highs.plot()
```

```python
df.drop(columns=['Symbol'],inplace=True)
```

```python
df.resample('W').mean()
```

```python
# but we dont need something like that
# eg we need weekly high instead of its mean
# and weekly low using min
# for close we may need mean
# for volume we need total amount like sum

# for these we are going to use aggergate function agg()

df.resample('W').agg({'Close':'mean','High':'max','Low':'min','Volume': 'sum'})
```

#  Reading/Writing Data to Different Sources - Excel, JSON, SQL, Etc

```python
import pandas as pd

df=pd.read_csv('data/results.csv',index_col='Respondent')
pd.set_option('display.max_columns',85)

# suppose u just wanna data's from india alone
filt = (df['Country'] == 'India')
modified = df.loc[filt]
```

```python
# to convert into csv and store
modified.to_csv('modified/modified_india.csv')
```

```python
# to convert into tsv and store-->tab separated values
modified.to_csv('modified/modified_india.tsv',sep='\t')
```

```python
# to create as excel file

modified.to_excel('modified/modified_india.xlsx')
```

```python
# to open the excel file

test = pd.read_excel('modified/modified_india.xlsx',index_col = 'Respondent')
```

```python
test
```

```python
# to create a json file
# it return as dictionary kinda data

modified.to_json('modified/modified_india.json')
```

```python
# if u need json as list like data instead of dictionary
modified.to_json('modified/modified_india.json', orient= 'records', lines=True)
```
