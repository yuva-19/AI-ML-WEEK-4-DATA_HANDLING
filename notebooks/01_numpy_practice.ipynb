Q3. Border Matrix ⭐⭐⭐

Create a 10×10 matrix

Border = 1
Inside = 0

Output

1111111111

1000000001

1000000001

...

1111111111

```python
mat=np.zeros((10,10),dtype='int8')
mat[0,:]=1
mat[-1,:]=1
mat[:,0]=1
mat[:,-1]=1
print(mat)
```

Q8. Conditional Replacement ⭐⭐⭐⭐

Given

a = np.array([2,-3,5,-1,8,-6])

Replace negative values with their absolute value.

```python
a = np.array([2,-3,5,-1,8,-6])
b = np.where(a<0,a*(-1),a)
print(b)
```

Distance Calculation ⭐⭐⭐⭐

Given

points = np.array([
[1,2],
[4,6],
[7,8]
])

Find the Euclidean distance of every point from the origin (0,0).

```python
points = np.array([
[1,2],
[4,6],
[7,8]
])

distance=np.sqrt(np.sum(points**2,axis=1))
print(distance)
```

# Section 6: Hard AI/ML Problems (Q19–Q20)

Q19. Min-Max Scaling ⭐⭐⭐⭐⭐

```python
X = np.array([
[100,200],
[150,300],
[120,250]
])

mini=np.min(X,axis=0)
maxi=np.max(X,axis=0)

X_scaled=(X-mini)/(maxi-mini)
print(X_scaled)
```

Q20. Outlier Detection

```python
data = np.array([
12,15,18,20,22,24,25,27,
29,30,35,120
])

Q1=np.percentile(data,25)
Q3=np.percentile(data,75)
IQR=Q3-Q1
print(IQR)
lower=Q1-1.5*(IQR)
upper=Q3+1.5*(IQR)
outliers=data[(data<lower) | (data>upper)]
print(outliers)
```
